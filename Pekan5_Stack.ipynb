# Pekan 5 - Stack

## Apa Itu Stack?

**Stack** adalah struktur data yang mengikuti prinsip **LIFO (Last In, First Out)** — elemen yang **terakhir dimasukkan** adalah yang **pertama dikeluarkan**.

Bayangkan seperti tumpukan piring: piring yang terakhir diletakkan di atas adalah yang pertama diambil.

```
        push(40)            pop()
           |                  ^
           v                  |
        +------+
TOP --> |  30  |
        +------+
        |  20  |
        +------+
        |  10  |    <-- dasar (bottom)
        +------+
```

Semua operasi hanya terjadi di **satu ujung**, yaitu **TOP**.

**Operasi utama Stack:**
- `push` → menambahkan elemen ke atas stack
- `pop` → mengambil dan menghapus elemen teratas
- `peek` → melihat elemen teratas tanpa menghapusnya
- `is_empty` → mengecek apakah stack kosong
- `size` → jumlah elemen
- `clear` → mengosongkan stack

Pada implementasi list Python, **akhir list adalah TOP**, sehingga `append()` dan `pop()` sama-sama O(1).

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP


s = Stack()
print(s.items)    # Output: []
````

## ✅ 2. Memeriksa Apakah Stack Kosong (`is_empty`)

### Konsep

Sebelum `pop` atau `peek`, kita perlu tahu apakah stack punya isi. Mengambil dari stack kosong akan menimbulkan error.

| Baris | Penjelasan |
|---|---|

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

    def is_empty(self):
        return len(self.items) == 0     # True jika stack kosong


s = Stack()
print(s.is_empty())    # Output: True
````

## ➕ 3. Menambahkan Elemen (`push`)

### Konsep

Menambahkan elemen baru ke **atas** stack. Elemen yang baru masuk menjadi TOP.

```
Sebelum push(30):      Sesudah push(30):

 TOP --> [20]           TOP --> [30]
         [10]                   [20]
                                [10]
```

| Baris | Penjelasan |
|---|---|
| `self.items.append(data)` | Menambah elemen di akhir list (= TOP) |

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

    def is_empty(self):
        return len(self.items) == 0     # True jika stack kosong

    def push(self, data):
        self.items.append(data)         # taruh di atas (akhir list = TOP)
        print(f"Push {data} ke stack")


s = Stack()
s.push(10)
s.push(20)
s.push(30)
print(s.items)    # Output: [10, 20, 30]
````

## ❌ 4. Mengambil Elemen Teratas (`pop`)

### Konsep

Mengambil dan **menghapus** elemen paling atas (prinsip LIFO). Elemen yang terakhir masuk adalah yang pertama keluar.

```
Sebelum pop():          Sesudah pop():

 TOP --> [30]            TOP --> [20]
         [20]                    [10]
         [10]            (30 diambil dan dihapus)
```

| Baris | Penjelasan |
|---|---|
| `if self.is_empty():` | Cek kosong dulu untuk menghindari error |
| `self.items.pop()` | Mengambil dan menghapus elemen terakhir (TOP) |
| `return dikeluarkan` | Mengembalikan nilai yang diambil agar bisa dipakai |

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

    def is_empty(self):
        return len(self.items) == 0     # True jika stack kosong

    def push(self, data):
        self.items.append(data)         # taruh di atas (akhir list = TOP)
        print(f"Push {data} ke stack")

    def pop(self):
        if self.is_empty():             # cek dulu supaya tidak error
            print("Stack kosong, tidak ada yang bisa di-pop!")
            return None
        dikeluarkan = self.items.pop()  # ambil elemen paling atas (LIFO)
        print(f"Pop: {dikeluarkan} dikeluarkan dari stack")
        return dikeluarkan


s = Stack()
s.push(10)
s.push(20)
s.push(30)
s.pop()            # Output: Pop: 30 dikeluarkan dari stack
s.pop()            # Output: Pop: 20 dikeluarkan dari stack
print(s.items)     # Output: [10]

kosong = Stack()
kosong.pop()       # Output: Stack kosong, tidak ada yang bisa di-pop!
````

## 🔍 5. Melihat Elemen Teratas (`peek`)

### Konsep

Melihat nilai TOP **tanpa menghapusnya**. Berbeda dengan `pop`, isi stack tidak berubah.

| Baris | Penjelasan |
|---|---|
| `self.items[-1]` | Index `-1` = elemen terakhir = TOP |
| Tidak ada `.pop()` | Hanya membaca, tidak menghapus |

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

    def is_empty(self):
        return len(self.items) == 0     # True jika stack kosong

    def push(self, data):
        self.items.append(data)

    def peek(self):
        if self.is_empty():
            print("Stack kosong!")
            return None
        return self.items[-1]           # elemen TOP, tanpa menghapusnya


s = Stack()
s.push(10)
s.push(20)
s.push(30)
print(s.peek())    # Output: 30
print(s.items)     # Output: [10, 20, 30]
````

## 🔢 6. Menghitung Jumlah Elemen (`size`)

### Konsep

Menghitung berapa banyak elemen di dalam stack.

| Baris | Penjelasan |
|---|---|
| `len(self.items)` | Jumlah elemen pada list |

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

    def push(self, data):
        self.items.append(data)

    def size(self):
        return len(self.items)


s = Stack()
s.push(10)
s.push(20)
s.push(30)
print("Jumlah elemen:", s.size())    # Output: Jumlah elemen: 3
````

## 🖥️ 7. Menampilkan Isi Stack (`display`)

### Konsep

Menampilkan stack dari **TOP ke bawah**, supaya tampilannya mirip tumpukan sebenarnya. Karena TOP ada di akhir list, kita membalik urutan dengan `reversed()`.

| Baris | Penjelasan |
|---|---|
| `reversed(self.items)` | Membaca list dari belakang (TOP) ke depan |
| `enumerate(...)` | Memberi nomor urut supaya elemen pertama (TOP) bisa diberi tanda |
| `"TOP --> "` | Penanda posisi TOP |

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

    def is_empty(self):
        return len(self.items) == 0     # True jika stack kosong

    def push(self, data):
        self.items.append(data)

    def display(self):
        if self.is_empty():
            print("Stack kosong")
            return
        print("Stack (dari TOP ke bawah):")
        for i, item in enumerate(reversed(self.items)):   # TOP dicetak paling atas
            awal = "TOP --> " if i == 0 else "        "
            print(f"{awal}| {item} |")
        print("        +----+")


s = Stack()
s.push(10)
s.push(20)
s.push(30)
s.display()

# Output yang diharapkan:
# Stack (dari TOP ke bawah):
# TOP --> | 30 |
#         | 20 |
#         | 10 |
#         +----+
````

## 🧹 8. Mengosongkan Stack (`clear`)

### Konsep

Menghapus semua elemen sekaligus.

| Baris | Penjelasan |
|---|---|
| `self.items = []` | Mengganti isi dengan list kosong |

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

    def is_empty(self):
        return len(self.items) == 0     # True jika stack kosong

    def push(self, data):
        self.items.append(data)

    def display(self):
        if self.is_empty():
            print("Stack kosong")
            return
        print("Stack (dari TOP ke bawah):")
        for i, item in enumerate(reversed(self.items)):   # TOP dicetak paling atas
            awal = "TOP --> " if i == 0 else "        "
            print(f"{awal}| {item} |")
        print("        +----+")

    def clear(self):
        self.items = []
        print("Stack berhasil dikosongkan")


s = Stack()
s.push(10)
s.push(20)
s.display()      # menampilkan 2 elemen
s.clear()        # Output: Stack berhasil dikosongkan
s.display()      # Output: Stack kosong
````

## 🔤 9. Aplikasi: Membalik String

### Konsep

Karena LIFO, huruf yang terakhir dimasukkan akan keluar pertama. Jadi kalau kita `push` semua huruf lalu `pop` semuanya, hasilnya adalah kata yang terbalik.

| Baris | Penjelasan |
|---|---|
| `for ch in teks: s.push(ch)` | Masukkan huruf satu per satu |
| `while not s.is_empty():` | Ulangi sampai stack kosong |
| `hasil += s.pop()` | Huruf terakhir yang masuk keluar lebih dulu |

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

    def is_empty(self):
        return len(self.items) == 0     # True jika stack kosong

    def push(self, data):
        self.items.append(data)

    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            print("Stack kosong!")
            return None
        return self.items[-1]           # elemen TOP, tanpa menghapusnya

    def size(self):
        return len(self.items)


def balik_string(teks):
    s = Stack()
    for ch in teks:
        s.push(ch)
    hasil = ""
    while not s.is_empty():
        hasil += s.pop()
    return hasil


print(balik_string("STRUKTUR"))    # Output: RUTKURTS
print(balik_string("Halo"))        # Output: olaH
````

## 🧮 10. Aplikasi: Mengecek Kurung Seimbang

### Konsep

Contoh klasik penggunaan stack. Setiap **kurung buka** di-`push`. Setiap **kurung tutup** harus cocok dengan kurung buka yang ada di TOP (hasil `pop`). Ekspresi seimbang jika semua cocok dan stack kosong di akhir.

```
Ekspresi: { [ ( ) ] }

 {   push {      stack: {
 [   push [      stack: { [
 (   push (      stack: { [ (
 )   pop  (  cocok
 ]   pop  [  cocok
 }   pop  {  cocok     stack kosong -> SEIMBANG
```

| Baris | Penjelasan |
|---|---|
| `pasangan = {")": "(", ...}` | Kamus pasangan kurung tutup dan buka |
| `if ch in "([{":` | Kurung buka: `push` |
| `s.pop() != pasangan[ch]` | Kurung tutup tidak cocok dengan TOP: tidak seimbang |
| `return s.is_empty()` | Jika masih ada sisa kurung buka, tidak seimbang |

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

    def is_empty(self):
        return len(self.items) == 0     # True jika stack kosong

    def push(self, data):
        self.items.append(data)

    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            print("Stack kosong!")
            return None
        return self.items[-1]           # elemen TOP, tanpa menghapusnya

    def size(self):
        return len(self.items)


def kurung_seimbang(ekspresi):
    pasangan = {")": "(", "]": "[", "}": "{"}
    s = Stack()
    for ch in ekspresi:
        if ch in "([{":
            s.push(ch)
        elif ch in ")]}":
            if s.is_empty() or s.pop() != pasangan[ch]:
                return False
    return s.is_empty()


print("{[()]} ->", kurung_seimbang("{[()]}"))    # Output: {[()]} -> True
print("([)]   ->", kurung_seimbang("([)]"))      # Output: ([)]   -> False
print("(((    ->", kurung_seimbang("((("))        # Output: (((    -> False
````

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

    def is_empty(self):
        return len(self.items) == 0     # True jika stack kosong

    def push(self, data):
        self.items.append(data)         # taruh di atas (akhir list = TOP)
        print(f"Push {data} ke stack")

    def pop(self):
        if self.is_empty():             # cek dulu supaya tidak error
            print("Stack kosong, tidak ada yang bisa di-pop!")
            return None
        dikeluarkan = self.items.pop()  # ambil elemen paling atas (LIFO)
        print(f"Pop: {dikeluarkan} dikeluarkan dari stack")
        return dikeluarkan

    def peek(self):
        if self.is_empty():
            print("Stack kosong!")
            return None
        return self.items[-1]           # elemen TOP, tanpa menghapusnya

    def size(self):
        return len(self.items)

    def display(self):
        if self.is_empty():
            print("Stack kosong")
            return
        print("Stack (dari TOP ke bawah):")
        for i, item in enumerate(reversed(self.items)):   # TOP dicetak paling atas
            awal = "TOP --> " if i == 0 else "        "
            print(f"{awal}| {item} |")
        print("        +----+")

    def clear(self):
        self.items = []
        print("Stack berhasil dikosongkan")


if __name__ == "__main__":
    s = Stack()

    print("=== Push ===")
    s.push(10)
    s.push(20)
    s.push(30)
    s.display()
    print("Jumlah elemen:", s.size())    # Output: Jumlah elemen: 3

    print("\n=== Peek ===")
    print("TOP saat ini:", s.peek())      # Output: TOP saat ini: 30

    print("\n=== Pop ===")
    s.pop()
    s.pop()
    s.display()

    print("\n=== Clear ===")
    s.clear()
    s.display()
````

## 🔑 Operasi Penting yang Wajib Dipahami

| Operasi | Cara Kerja | Kompleksitas |
|---|---|---|
| `push(data)` | `items.append(data)`, tambah ke TOP | O(1) |
| `pop()` | `items.pop()`, hapus dari TOP | O(1) |
| `peek()` | `items[-1]`, baca TOP tanpa hapus | O(1) |
| `is_empty()` | `len(items) == 0` | O(1) |
| `size()` | `len(items)` | O(1) |
| `clear()` | `items = []` | O(1) |

**Contoh penggunaan stack:** tombol *undo* di editor, riwayat *back* di browser, mengecek kurung seimbang, membalik string, mengevaluasi ekspresi postfix, dan pemanggilan fungsi (call stack).

> 🧠 **Tips kunci:** Stack bekerja dengan prinsip LIFO: `push` dan `pop` selalu di **TOP**. Gunakan `peek()` jika hanya ingin melihat TOP tanpa mengubah isinya. Selalu cek `is_empty()` sebelum `pop()` atau `peek()` untuk menghindari error!

## 🏋️ Soal Latihan

### Soal 1 - Evaluasi Ekspresi Postfix (`evaluasi_postfix`)

**Deskripsi:**
Buatlah fungsi `evaluasi_postfix(ekspresi)` untuk menghitung ekspresi **postfix** (operator ditulis setelah operand) menggunakan Stack. Aturannya: jika token berupa **angka**, `push` ke stack. Jika token berupa **operator**, `pop` dua angka (angka **kanan** `b` diambil lebih dulu, lalu angka kiri `a`), hitung `a (operator) b`, lalu `push` hasilnya. Hasil akhir adalah satu-satunya isi stack.

**Contoh:**
```
"3 4 + 2 *"            ->  (3 + 4) * 2           = 14
"5 6 2 + * 12 4 / -"   ->  5 * (6 + 2) - 12 / 4  = 37
```

**Petunjuk:**
Untuk pembagian gunakan `//` supaya hasilnya bilangan bulat. Perhatikan urutan: `b = s.pop()` dulu, baru `a = s.pop()`, sebab `a - b` berbeda dengan `b - a`.

```python
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

            # TODO 2: hitung a (token) b lalu push hasilnya ke stack
            pass
        else:
            s.push(int(token))

```

```python
# ===== MAIN =====
```

````python
class Stack:
    def __init__(self):
        self.items = []                 # list kosong sebagai wadah; ujung list = TOP

    def is_empty(self):
        return len(self.items) == 0     # True jika stack kosong

    def push(self, data):
        self.items.append(data)

    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    def peek(self):
        if self.is_empty():
            print("Stack kosong!")
            return None
        return self.items[-1]           # elemen TOP, tanpa menghapusnya

    def size(self):
        return len(self.items)


def evaluasi_postfix(ekspresi):
    s = Stack()
    for token in ekspresi.split():
        if token in "+-*/":
            b = s.pop()                       # TODO 1
            a = s.pop()
            if token == "+":                  # TODO 2
                s.push(a + b)
            elif token == "-":
                s.push(a - b)
            elif token == "*":
                s.push(a * b)
            else:
                s.push(a // b)
        else:
            s.push(int(token))

    return s.pop()                            # TODO 3


# ===== MAIN =====
print(evaluasi_postfix("3 4 + 2 *"))
print(evaluasi_postfix("5 6 2 + * 12 4 / -"))

# Output yang diharapkan:
# 14
# 37
````
