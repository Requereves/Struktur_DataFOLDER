# Pekan 1 - Pointer dan Abstract Data Type (ADT)

## Apa Itu Pointer dan ADT?

### Pointer (Referensi)

**Pointer** adalah variabel yang menyimpan **alamat memori** dari suatu data, bukan data itu sendiri. Di C/C++ pointer ditulis eksplisit (`int *p`). Di **Python tidak ada sintaks pointer**, tetapi hasilnya sama: **setiap variabel adalah referensi (penunjuk) ke sebuah objek** di memori.

```
a = [1, 2, 3]          b = a

a --> [1, 2, 3]        a --\
                            +--> [1, 2, 3]     (satu objek, dua nama)
                       b --/
```

Hal penting yang harus diingat:
- `b = a` **tidak menyalin** data, hanya membuat `b` menunjuk ke objek yang sama dengan `a`.
- `None` adalah **null pointer**: variabel yang tidak menunjuk ke objek apa pun.
- Konsep ini adalah fondasi **linked list**, karena setiap node menunjuk ke node berikutnya lewat pointer `next`.

### Abstract Data Type (ADT)

**ADT** adalah tipe data yang didefinisikan oleh:
1. **Data** yang disimpan, dan
2. **Operasi** yang boleh dilakukan pada data tersebut,

tanpa memperlihatkan cara kerja di dalamnya (implementasi). Pengguna cukup tahu *apa* yang bisa dilakukan, bukan *bagaimana* caranya.

Contoh ADT yang akan sering dipakai: `List`, `Stack`, `Queue`. Di Python, ADT dibuat dengan `class`.

## 📌 1. Variabel adalah Referensi

### Konsep

Nama variabel di Python hanyalah **label** yang menunjuk ke sebuah objek. Saat menulis `b = a`, Python tidak membuat salinan data, tetapi membuat `b` menunjuk ke **objek yang sama** dengan `a`. Akibatnya, perubahan lewat `b` juga terlihat lewat `a`.

```
Sebelum append:            Sesudah b.append(4):

a --\                      a --\
     +--> [1, 2, 3]              +--> [1, 2, 3, 4]
b --/                      b --/
```

| Baris | Penjelasan |
|---|---|
| `b = a` | `b` menunjuk objek yang sama dengan `a` (bukan salinan) |
| `b.append(4)` | Mengubah objek asli, sehingga `a` ikut berubah |
| `id(x)` | Mengembalikan identitas/alamat objek |
| `a is b` | `True` jika `a` dan `b` adalah objek yang **sama** |

````python
a = [1, 2, 3]
b = a                      # b TIDAK menyalin, b menunjuk objek yang sama
b.append(4)

print("a =", a)            # Output: a = [1, 2, 3, 4]
print("b =", b)            # Output: b = [1, 2, 3, 4]
print("Alamat sama?", id(a) == id(b))   # Output: Alamat sama? True
print("a is b?", a is b)   # Output: a is b? True
````

## 📋 2. Alias vs Salinan (`copy` dan `deepcopy`)

### Konsep

Kalau ingin data yang **benar-benar terpisah**, kita harus membuat salinan. Ada tiga kondisi:

- **Alias**: `alias = asli` (objek sama).
- **Salinan dangkal** (`copy()`): lapisan luar disalin, tetapi objek di dalamnya (misal list di dalam list) **masih dipakai bersama**.
- **Salinan dalam** (`copy.deepcopy()`): semua lapisan disalin, benar-benar terpisah.

| Baris | Penjelasan |
|---|---|
| `import copy` | Memuat modul `copy` untuk `deepcopy` |
| `alias = asli` | Hanya nama baru untuk objek yang sama |
| `asli.copy()` | Salinan dangkal (shallow copy) |
| `copy.deepcopy(asli)` | Salinan penuh (deep copy) |
| `asli[0].append(99)` | Mengubah list **di dalam** list asli |

````python
import copy

asli = [[1, 2], [3, 4]]
alias = asli                       # alias: objek sama
dangkal = asli.copy()              # salinan dangkal
dalam = copy.deepcopy(asli)        # salinan dalam

asli[0].append(99)                 # ubah list di DALAM list

print("asli    :", asli)          # Output: asli    : [[1, 2, 99], [3, 4]]
print("alias   :", alias)         # Output: alias   : [[1, 2, 99], [3, 4]]
print("dangkal :", dangkal)       # Output: dangkal : [[1, 2, 99], [3, 4]]
print("dalam   :", dalam)         # Output: dalam   : [[1, 2], [3, 4]]
````

## 🚫 3. `None` sebagai Null Pointer

### Konsep

`None` berarti **tidak menunjuk ke apa pun**. Ini dipakai untuk menandai "kosong" atau "ujung". Di linked list, node terakhir memiliki `next = None`, dan list kosong memiliki `head = None`.

Selalu bandingkan dengan `is None` / `is not None` (bukan `== None`).

| Baris | Penjelasan |
|---|---|
| `p = None` | `p` tidak menunjuk ke objek apa pun |
| `p is None` | Cek apakah `p` kosong |
| `p is not None` | Cek apakah `p` menunjuk ke sesuatu |

````python
p = None
print("p kosong?", p is None)      # Output: p kosong? True

data = [10, 20]
p = data                           # sekarang p menunjuk ke list
print("p kosong?", p is None)      # Output: p kosong? False
````

## 🔗 4. Objek yang Menunjuk Objek Lain

### Konsep

Atribut sebuah objek juga bisa berupa referensi ke objek lain. Inilah cikal bakal **linked list**: setiap `Node` menyimpan `data` dan `next`, yaitu pointer ke node berikutnya.

```
n1             n2             n3
[ 10 | * ]--->[ 20 | * ]--->[ 30 | None ]
```

| Baris | Penjelasan |
|---|---|
| `self.data = data` | Nilai yang disimpan node |
| `self.next = None` | Pointer ke node berikutnya, awalnya kosong |
| `n1.next = n2` | `n1` sekarang menunjuk ke `n2` |
| `cur = cur.next` | Pindah ke node berikutnya (menelusuri rantai) |

````python
class Node:
    def __init__(self, data):
        self.data = data         # nilai yang disimpan
        self.next = None         # pointer ke node berikutnya

n1 = Node(10)
n2 = Node(20)
n3 = Node(30)
n1.next = n2                     # n1 menunjuk n2
n2.next = n3                     # n2 menunjuk n3

cur = n1
while cur is not None:           # telusuri sampai ujung (None)
    print(cur.data, end=" ")
    cur = cur.next
print()
# Output: 10 20 30
````

## 🏗️ 5. Membuat ADT: Atribut

### Konsep

ADT dibuat dengan `class`. Langkah pertama adalah menentukan **data** apa yang disimpan. Sebagai contoh kita buat ADT `Mahasiswa` dengan atribut `nim`, `nama`, dan `nilai`.

| Baris | Penjelasan |
|---|---|
| `class Mahasiswa:` | Cetak biru (blueprint) untuk ADT Mahasiswa |
| `def __init__(self, ...)` | Konstruktor, otomatis dipanggil saat `Mahasiswa(...)` dibuat |
| `self.nim = nim` | Menyimpan data ke dalam objek |
| `nilai=0` | Parameter default: jika tidak diisi, nilainya 0 |

````python
class Mahasiswa:
    def __init__(self, nim, nama, nilai=0):
        self.nim = nim
        self.nama = nama
        self.nilai = nilai

m = Mahasiswa("2301001", "Andi", 80)
print(m.nim, m.nama, m.nilai)    # Output: 2301001 Andi 80
````

## ⚙️ 6. Menambah Operasi pada ADT

### Konsep

ADT bukan hanya data, tetapi juga **operasi**. Operasi dibuat sebagai method. Di sini kita tambahkan `set_nilai` (dengan validasi), `get_grade`, dan `tampil`. Pengguna cukup memanggil method tanpa perlu tahu isi perhitungannya.

| Baris | Penjelasan |
|---|---|
| `set_nilai(nilai)` | Mengubah nilai hanya jika berada di rentang 0 sampai 100 |
| `get_grade()` | Mengembalikan grade A/B/C/D berdasarkan nilai |
| `tampil()` | Mencetak data mahasiswa dalam satu baris |
| `self.get_grade()` | Method boleh memanggil method lain di kelas yang sama |

````python
class Mahasiswa:
    def __init__(self, nim, nama, nilai=0):
        self.nim = nim
        self.nama = nama
        self.nilai = nilai

    def set_nilai(self, nilai):
        if 0 <= nilai <= 100:            # validasi: nilai harus 0 sampai 100
            self.nilai = nilai
        else:
            print("Nilai harus 0-100")

    def get_grade(self):
        if self.nilai >= 85:
            return "A"
        if self.nilai >= 70:
            return "B"
        if self.nilai >= 55:
            return "C"
        return "D"

    def tampil(self):
        print(f"{self.nim} | {self.nama} | {self.nilai} | {self.get_grade()}")


m = Mahasiswa("2301001", "Andi", 80)
m.tampil()            # Output: 2301001 | Andi | 80 | B
m.set_nilai(95)
m.tampil()            # Output: 2301001 | Andi | 95 | A
m.set_nilai(150)      # Output: Nilai harus 0-100
````

## 🔒 7. Enkapsulasi: Menyembunyikan Data

### Konsep

Salah satu inti ADT adalah **menyembunyikan implementasi**. Atribut yang diawali `__` (dua garis bawah) bersifat **privat**: tidak bisa diakses langsung dari luar kelas, hanya lewat method. Dengan begitu data tidak bisa diubah sembarangan (misalnya saldo menjadi negatif).

| Baris | Penjelasan |
|---|---|
| `self.__saldo` | Atribut privat, hanya boleh dipakai di dalam kelas |
| `get_saldo()` | Method untuk **membaca** saldo (getter) |
| `setor(jumlah)` | Menambah saldo dengan validasi jumlah > 0 |
| `tarik(jumlah)` | Mengurangi saldo hanya jika saldo cukup |
| `r.__saldo` | Mengakses langsung dari luar akan menimbulkan `AttributeError` |

````python
class Rekening:
    def __init__(self, pemilik, saldo=0):
        self.pemilik = pemilik
        self.__saldo = saldo             # atribut privat

    def get_saldo(self):
        return self.__saldo

    def setor(self, jumlah):
        if jumlah > 0:
            self.__saldo += jumlah
        else:
            print("Jumlah setor harus lebih dari 0")

    def tarik(self, jumlah):
        if jumlah > self.__saldo:
            print("Saldo tidak cukup")
        else:
            self.__saldo -= jumlah


r = Rekening("Budi", 100000)
r.setor(50000)
print(r.get_saldo())        # Output: 150000
r.tarik(200000)             # Output: Saldo tidak cukup
print(r.get_saldo())        # Output: 150000

try:
    print(r.__saldo)        # akses langsung ke atribut privat
except AttributeError:
    print("Tidak bisa diakses langsung")   # Output: Tidak bisa diakses langsung
````

## 🔑 Konsep Penting yang Wajib Dipahami

| Konsep | Arti | Contoh |
|---|---|---|
| `==` | Membandingkan **isi** | `[1] == [1]` → `True` |
| `is` | Membandingkan **objek yang sama** | `a is b` |
| `id(x)` | Identitas/alamat objek | `id(a)` |
| `copy()` | Salinan dangkal | `b = a.copy()` |
| `deepcopy()` | Salinan penuh | `copy.deepcopy(a)` |
| `None` | Null pointer | `p = None` |
| `self.__x` | Atribut privat (enkapsulasi) | `self.__saldo` |
| ADT | Data + operasi, implementasi disembunyikan | `Mahasiswa`, `Rekening` |

> 🧠 **Tips kunci:** `b = a` itu **alias**, bukan salinan. Jika ingin data terpisah, gunakan `copy()` (untuk list sederhana) atau `deepcopy()` (untuk list di dalam list). Gunakan `is None` untuk mengecek pointer kosong, dan jangan lupa bahwa ADT yang baik hanya membuka **operasi**, bukan cara kerjanya.

````python
from math import gcd


class Pecahan:
    def __init__(self, pembilang, penyebut):
        if penyebut == 0:
            raise ValueError("Penyebut tidak boleh 0")
        self.pembilang = pembilang
        self.penyebut = penyebut

    def sederhanakan(self):
        f = gcd(self.pembilang, self.penyebut)      # TODO 1
        self.pembilang //= f
        self.penyebut //= f

    def tambah(self, lain):
        pembilang = self.pembilang * lain.penyebut + lain.pembilang * self.penyebut   # TODO 2
        penyebut = self.penyebut * lain.penyebut
        hasil = Pecahan(pembilang, penyebut)
        hasil.sederhanakan()
        return hasil

    def tampil(self):
        print(f"{self.pembilang}/{self.penyebut}")   # TODO 3


# ===== MAIN =====
a = Pecahan(1, 2)
b = Pecahan(1, 3)
a.tambah(b).tampil()

c = Pecahan(6, 8)
c.sederhanakan()
c.tampil()

# Output yang diharapkan:
# 5/6
# 3/4
````
