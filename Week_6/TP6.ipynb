## Apa Itu Stack?

**Stack** adalah struktur data yang mengikuti prinsip **LIFO (Last In, First Out)** — elemen yang **terakhir dimasukkan** adalah yang **pertama dikeluarkan**.

Bayangkan seperti tumpukan piring: piring yang ditaruh paling atas adalah yang pertama diambil.

```
       [  30  ]  <-- TOP (elemen terakhir masuk, pertama keluar)
       [  20  ]
       [  10  ]  <-- BOTTOM
       --------
```

Setiap **stack** terdiri dari:
- `data` → nilai yang disimpan di setiap elemen
- `top` → pointer/indeks yang selalu menunjuk ke elemen **paling atas**

**Operasi utama Stack:**
- `push` → menambahkan elemen ke atas stack
- `pop` → mengambil dan menghapus elemen teratas
- `peek` → melihat elemen teratas tanpa menghapusnya
- `is_empty` → mengecek apakah stack kosong

**`top`** selalu menunjuk ke elemen yang paling atas (terakhir dimasukkan).  
Jika stack kosong, biasanya `top = -1` (implementasi array) atau `top = None` (implementasi linked list).

## ➕ 3. Menambahkan Elemen (`push`)

### Konsep
Menambahkan elemen **baru** ke posisi **paling atas** stack. Elemen yang baru ditambahkan langsung menjadi `top`.

Langkah-langkahnya:
1. Ambil elemen baru
2. Letakkan di atas stack (akhir list)
3. Elemen baru menjadi top

```
Sebelum push(30):        Sesudah push(30):
  [ 20 ]  <-- top          [ 30 ]  <-- top (baru)
  [ 10 ]                   [ 20 ]
  ------                   [ 10 ]
                            ------
```

```python
class Stack:
    def __init__(self):
        self.items = []
    
    def push(self, data):
        self.items.append(data)      # Tambahkan elemen ke atas stack (akhir list)
        print(f"Push {data} ke stack")
```

| Baris | Penjelasan |
|---|---|
| `self.items.append(data)` | Menambahkan elemen baru ke bagian atas (akhir list = top stack) |

```python
s = Stack()
```

## ❌ 4. Mengambil Elemen Teratas (`pop`)

### Konsep
Mengambil dan **menghapus** elemen yang berada di **paling atas** stack (prinsip LIFO). Elemen yang terakhir masuk adalah yang pertama keluar.

```
Sebelum pop():              Sesudah pop():
  [ 30 ]  <-- top (diambil)   [ 20 ]  <-- top (baru)
  [ 20 ]                       [ 10 ]
  [ 10 ]                       ------
  ------
```

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, data):
        self.items.append(data)

    def pop(self):
        if self.is_empty():              # Cek dulu apakah stack kosong
            print("Stack kosong, tidak ada yang bisa di-pop!")
            return None
        popped = self.items.pop()        # Ambil dan hapus elemen teratas
        print(f"Pop: {popped} dikeluarkan dari stack")
        return popped                    # Kembalikan nilai yang diambil

    def is_empty(self):
        return len(self.items) == 0
```

| Baris | Penjelasan |
|---|---|
| `if self.is_empty():` | Selalu cek kosong sebelum pop untuk menghindari error |
| `self.items.pop()` | Mengambil dan menghapus elemen terakhir (top stack) |
| `return popped` | Mengembalikan nilai yang diambil agar bisa dipakai |

```python
s = Stack()

s.push(20)
s.push(30)
s.pop()        # Output: Pop: 30 dikeluarkan dari stack
```

## 🔍 5. Melihat Elemen Teratas (`peek`)

### Konsep
Melihat nilai elemen **paling atas** stack **tanpa menghapusnya**. Berbeda dengan `pop`, elemen tetap ada di stack setelah `peek`.

```
Stack saat ini:   peek() mengembalikan 30, stack TIDAK berubah:
  [ 30 ]  <--       [ 30 ]  <-- top (masih ada)
  [ 20 ]            [ 20 ]
  [ 10 ]            [ 10 ]
  ------            ------
```

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, data):
        self.items.append(data)

    def is_empty(self):
        return len(self.items) == 0

    def peek(self):
        if self.is_empty():              # Cek apakah stack kosong
            print("Stack kosong!")
            return None
        return self.items[-1]            # Kembalikan elemen teratas TANPA menghapusnya
```

| Baris | Penjelasan |
|---|---|
| `self.items[-1]` | Indeks `-1` di Python berarti elemen terakhir (paling atas/top stack) |
| Tidak ada `.pop()` | Hanya membaca nilai, tidak menghapus |

```python
s = Stack()
s.push(10)
s.push(20)
s.push(30)
print(s.peek())    # Output: 30
print(s.items)     # Output: [10, 20, 30]  (tidak berubah)
```

## 🖥️ 7. Menampilkan Isi Stack (`display`)

### Konsep
Menampilkan seluruh elemen stack dari **atas (top) ke bawah (bottom)**, sehingga terlihat seperti tumpukan yang sesungguhnya.

```python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, data):
        self.items.append(data)

    def is_empty(self):
        return len(self.items) == 0

    def display(self):
        if self.is_empty():              # Cek apakah stack kosong
            print("Stack kosong")
            return

        print("Stack (dari atas ke bawah):")
        for item in reversed(self.items):    # Iterasi dari belakang ke depan (top ke bottom)
            print(f"  [ {item} ]")
        print("  ------")
```

| Baris | Penjelasan |
|---|---|
| `reversed(self.items)` | Membalik urutan list agar top tampil duluan |
| `for item in ...` | Kunjungi setiap elemen dari atas ke bawah |


````python
# ===================== DEMO =====================
class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, data):
        self.items.append(data)
        print(f"Push {data} ke stack")

    def pop(self):
        if self.is_empty():
            print("Stack kosong, tidak ada yang bisa di-pop!")
            return None
        popped = self.items.pop()
        print(f"Pop: {popped} dikeluarkan dari stack")
        return popped

    def peek(self):
        if self.is_empty():
            print("Stack kosong!")
            return None
        return self.items[-1]

    def size(self):
        return len(self.items)

    def display(self):
        if self.is_empty():
            print("Stack kosong")
            return
        print("Stack (dari atas ke bawah):")
        for item in reversed(self.items):
            print(f"  [ {item} ]")
        print("  ------")

    def clear(self):
        self.items = []
        print("Stack berhasil dikosongkan")


if __name__ == "__main__":
    s = Stack()

    print("=== Push ===")
    s.push(10)
    s.push(20)
    s.push(30)
    s.push(40)
    s.display()
    print("Jumlah elemen:", s.size())    # 4

    print("\n=== Peek ===")
    print("Top saat ini:", s.peek())     # 40

    print("\n=== Pop ===")
    s.pop()
    s.pop()
    s.display()

    print("\n=== Clear ===")
    s.clear()
    s.display()
````

## 📐 Konsep Stack: Ringkasan Visual

```
KONDISI AWAL (stack kosong):
items = []

SETELAH push(10):
items = [10]
         ^
        TOP

SETELAH push(20):
items = [10, 20]
              ^
             TOP

SETELAH push(30):
items = [10, 20, 30]
                  ^
                 TOP

SAAT pop():
  Ambil items[-1] = 30  →  kembalikan 30
  items = [10, 20]
                ^
               TOP baru

SAAT peek():
  Baca items[-1] = 20  →  kembalikan 20 (items TIDAK berubah)
  items = [10, 20]
```

## 🔑 Operasi Penting yang Wajib Dipahami

| Operasi | Cara Kerja | Kompleksitas |
|---|---|---|
| `push(data)` | `items.append(data)` — tambah ke akhir list | O(1) |
| `pop()` | `items.pop()` — hapus dari akhir list | O(1) |
| `peek()` | `items[-1]` — baca akhir list tanpa hapus | O(1) |
| `is_empty()` | `len(items) == 0` — cek kosong | O(1) |
| `size()` | `len(items)` — hitung elemen | O(1) |
| `clear()` | `items = []` — reset stack | O(1) |

> 🧠 **Tips kunci:** Stack bekerja dengan prinsip LIFO — selalu ingat bahwa operasi `push` dan `pop` hanya terjadi di **satu ujung** (TOP). Gunakan `peek()` jika hanya ingin melihat top tanpa mengubah isi stack. Selalu cek `is_empty()` sebelum `pop()` atau `peek()` untuk menghindari error!

## 🏋️ Soal Latihan

### Soal 1 — Cek Tanda Kurung Seimbang (`is_balanced`)

**Deskripsi:**  
Buatlah fungsi `is_balanced(expression)` yang menerima sebuah string ekspresi matematika dan mengembalikan `True` jika semua tanda kurung `(`, `)` dalam ekspresi tersebut **seimbang** (setiap kurung buka punya pasangan kurung tutup), atau `False` jika tidak seimbang. Gunakan Stack untuk menyelesaikan soal ini.

**Contoh:**
```
Input : "(3 + 5) * (2 - 1)"
Output: True

Input : "((4 + 2) * 3"
Output: False

Input : ")(1 + 2)"
Output: False
```

````python
class Stack:
    def __init__(self):
        self.items = []

    def push(self, data):
        self.items.append(data)

    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    def is_empty(self):
        return len(self.items) == 0


def is_balanced(expression):
    s = Stack()                          # Buat stack baru

    for char in expression:              # Iterasi setiap karakter
        if char == '(':                  # Jika kurung buka...
            # TODO 1: push karakter ke stack
            s.push(char)
            # pass

        elif char == ')':                # Jika kurung tutup...
            if s.is_empty():             # Jika stack kosong, tidak ada pasangannya
                # TODO 2: return False (tidak seimbang)
                 return False
                # pass
            # TODO 3: pop dari stack (buang pasangan kurung buka)
            s.pop()

    # TODO 4: Setelah loop selesai, kembalikan True jika stack kosong, False jika tidak
    return s.is_empty()


# ===== MAIN =====
ekspresi1 = "(3 + 5) * (2 - 1)"
ekspresi2 = "((4 + 2) * 3"
ekspresi3 = ")(1 + 2)"

print(f'"{ekspresi1}" → Seimbang: {is_balanced(ekspresi1)}')
print(f'"{ekspresi2}" → Seimbang: {is_balanced(ekspresi2)}')
print(f'"{ekspresi3}" → Seimbang: {is_balanced(ekspresi3)}')

# Output yang diharapkan:
# "(3 + 5) * (2 - 1)" → Seimbang: True
# "((4 + 2) * 3" → Seimbang: False
# ")(1 + 2)" → Seimbang: False
````
