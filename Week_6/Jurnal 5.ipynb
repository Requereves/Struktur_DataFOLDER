*Deskripsi:*  
Buatlah method `remove_duplicates(self)` yang menghapus semua node yang memiliki nilai *duplikat* (sama), hanya menyisakan kemunculan *pertama* dari setiap nilai. Double linked list tidak dijamin terurut.

*Contoh:*

```
Input : None <-- 1 <--> 3 <--> 2 <--> 3 <--> 5 <--> 1 <--> 2 --> None
Output: None <-- 1 <--> 3 <--> 2 <--> 5 --> None
```

*Petunjuk:*  
Gunakan sebuah `set` Python untuk menyimpan nilai yang sudah pernah ditemui. Lakukan traversal; jika nilai node sudah ada di set, hapus node tersebut dengan menyambungkan `current.prev.next` ke `current.next` **dan** `current.next.prev` ke `current.prev` (jangan lupa update `tail` jika node yang dihapus adalah node terakhir). Jika belum ada, tambahkan ke set dan lanjutkan.

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None
        self.prev = None


class DoubleLinkedList:
    def __init__(self):
        self.head = None
        self.tail = None

    def insert_at_end(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            self.tail = new_node
            return
        new_node.prev = self.tail
        self.tail.next = new_node
        self.tail = new_node

    def display(self):
        if self.head is None:
            print("Double linked list kosong")
            return
        current = self.head
        result = []
        while current is not None:
            result.append(str(current.data))
            current = current.next
        print("None <-- " + " <--> ".join(result) + " --> None")

    def remove_duplicates(self):
        if self.head is None:   # Jika list kosong, langsung keluar
            return

        seen = set()            # TODO 1: Set untuk menyimpan nilai yang sudah ditemui
        current = self.head     # Mulai dari head

        while current is not None:
            if current.data in seen:
                # TODO 2: Data sudah pernah ada → hapus node ini
                # Sambungkan node sebelumnya langsung ke node sesudah current:
                current.prev.next = current.next

                # TODO 3: Jika current BUKAN tail, sambungkan node sesudahnya balik ke sebelum current:
                if current.next is not None:
                    current.next.prev = current.prev
                # Jika current ADALAH tail, update self.tail:
                else:
                    self.tail = current.prev

                # pass            # Hapus baris ini setelah mengisi TODO 2 dan 3
            else:
                # TODO 4: Data belum pernah ada → tambahkan ke seen
                seen.add(current.data)
                # pass            # Hapus baris ini setelah mengisi TODO 4

            # TODO 5: Maju ke node berikutnya
            current = current.next
````

````python
# ===== MAIN =====
dll = DoubleLinkedList()
for val in [1, 3, 2, 3, 5, 1, 2]:
    dll.insert_at_end(val)

print("Sebelum remove_duplicates:")
dll.display()

dll.remove_duplicates()

print("Sesudah remove_duplicates:")
dll.display()

# Output yang diharapkan:
# Sebelum remove_duplicates:
# None <-- 1 <--> 3 <--> 2 <--> 3 <--> 5 <--> 1 <--> 2 --> None
# Sesudah remove_duplicates:
# None <-- 1 <--> 3 <--> 2 <--> 5 --> None
````
