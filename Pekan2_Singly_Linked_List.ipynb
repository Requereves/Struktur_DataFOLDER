## Apa Itu Singly Linked List?

**Singly Linked List** adalah struktur data berantai: kumpulan **node** yang saling terhubung **searah**. Setiap node menyimpan dua hal:
- `data` → nilai yang disimpan
- `next` → pointer ke node berikutnya (node terakhir menunjuk `None`)

```
head
 |
 v
[ 10 | * ]---->[ 20 | * ]---->[ 30 | None ]
  data  next     data  next     data   next
```

List hanya punya satu pintu masuk, yaitu **`head`** (node pertama). Untuk mencapai node lain, kita harus berjalan dari `head` mengikuti `next`.

**Dibanding list/array biasa:**
- Sisip/hapus di depan sangat cepat karena tidak perlu menggeser elemen.
- Ukuran fleksibel, tidak perlu menentukan kapasitas di awal.
- Tidak bisa langsung loncat ke index tertentu; harus berjalan dari `head`.

**Operasi yang dibahas di notebook ini:**
- `is_empty` → mengecek apakah list kosong
- `display` → menampilkan isi list
- `insert_front` → menambah node di depan
- `insert_last` → menambah node di belakang
- `length` → menghitung jumlah node
- `search` → mencari posisi sebuah nilai

**`head`** selalu menunjuk node pertama. Jika list kosong, `head = None`.

## 📦 1. Membuat Node

### Konsep

**Node** adalah satu "gerbong" pada rantai. Ia menyimpan `data` dan `next`. Saat baru dibuat, `next` masih `None` karena belum terhubung ke node lain.

| Baris | Penjelasan |
|---|---|
| `class Node:` | Cetak biru untuk satu node |
| `self.data = data` | Menyimpan nilai |
| `self.next = None` | Pointer ke node berikutnya, awalnya kosong |

````python
class Node:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.next = None          # pointer ke node berikutnya (awalnya kosong)


n1 = Node(10)
print(n1.data)    # Output: 10
print(n1.next)    # Output: None
````

## ✅ 3. Memeriksa Apakah List Kosong (`is_empty`)

### Konsep

List dianggap kosong jika `head` masih `None`. Cek ini penting sebelum menghapus atau menampilkan data.

| Baris | Penjelasan |
|---|---|
| `self.head is None` | `True` jika tidak ada node sama sekali |

````python
class Node:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.next = None          # pointer ke node berikutnya (awalnya kosong)


class SinglyLinkedList:
    def __init__(self):
        self.head = None              # list kosong: head menunjuk ke None

    def is_empty(self):
        return self.head is None      # True jika belum ada node


sll = SinglyLinkedList()
print(sll.is_empty())    # Output: True

sll.head = Node(10)
print(sll.is_empty())    # Output: False
````

## 🖥️ 4. Menampilkan Isi List (`display`)

### Konsep

Untuk menampilkan, kita **telusuri (traverse)** list: mulai dari `head`, cetak data, lalu pindah ke `next`, terus sampai bertemu `None`. Pola `while cur is not None` ini dipakai di hampir semua operasi linked list.

| Baris | Penjelasan |
|---|---|
| `cur = self.head` | Mulai dari node pertama |
| `while cur is not None:` | Ulangi selama masih ada node |
| `hasil.append(str(cur.data))` | Kumpulkan data setiap node |
| `cur = cur.next` | Pindah ke node berikutnya |
| `" -> ".join(hasil)` | Menggabungkan data dengan tanda panah |

````python
class Node:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.next = None          # pointer ke node berikutnya (awalnya kosong)


class SinglyLinkedList:
    def __init__(self):
        self.head = None              # list kosong: head menunjuk ke None

    def is_empty(self):
        return self.head is None      # True jika belum ada node

    def display(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head
        hasil = []
        while cur is not None:        # telusuri dari head sampai None
            hasil.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(hasil) + " -> None")


sll = SinglyLinkedList()
sll.display()                    # Output: List kosong

sll.head = Node(10)              # buat rantai manual: 10 -> 20 -> 30
sll.head.next = Node(20)
sll.head.next.next = Node(30)
sll.display()                    # Output: 10 -> 20 -> 30 -> None
````

## ➕ 5. Menambah di Depan (`insert_front`)

### Konsep

Menambahkan node **baru** sebagai `head`. Operasi ini cepat (O(1)) karena tidak perlu menelusuri list.

Langkah-langkahnya (urutan **penting**):
1. Buat node baru
2. Sambungkan `next` node baru ke `head` yang lama
3. Pindahkan `head` ke node baru

```
Sebelum insert_front(5):        Sesudah insert_front(5):

head --> [10|*]--> [20|None]    head --> [5|*]--> [10|*]--> [20|None]
```

| Baris | Penjelasan |
|---|---|
| `baru = Node(data)` | Membuat node baru |
| `baru.next = self.head` | Node baru menunjuk node pertama yang lama |
| `self.head = baru` | `head` pindah ke node baru |

````python
class Node:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.next = None          # pointer ke node berikutnya (awalnya kosong)


class SinglyLinkedList:
    def __init__(self):
        self.head = None              # list kosong: head menunjuk ke None

    def is_empty(self):
        return self.head is None      # True jika belum ada node

    def display(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head
        hasil = []
        while cur is not None:        # telusuri dari head sampai None
            hasil.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(hasil) + " -> None")

    def insert_front(self, data):
        baru = Node(data)             # 1. buat node baru
        baru.next = self.head         # 2. node baru menunjuk head lama
        self.head = baru              # 3. head pindah ke node baru


sll = SinglyLinkedList()
sll.insert_front(30)
sll.insert_front(20)
sll.insert_front(10)
sll.display()        # Output: 10 -> 20 -> 30 -> None
````

## ➕ 6. Menambah di Belakang (`insert_last`)

### Konsep

Menambahkan node baru di **akhir** list. Karena kita hanya punya `head`, kita harus berjalan dulu sampai node terakhir (node yang `next`-nya `None`), baru menyambungkan node baru. Karena itu operasinya O(n).

Ada dua kasus:
1. List kosong → node baru langsung menjadi `head`
2. List berisi → telusuri sampai node terakhir lalu sambungkan

| Baris | Penjelasan |
|---|---|
| `if self.is_empty():` | Kasus khusus: list kosong |
| `while cur.next is not None:` | Berhenti di node **terakhir** (bukan sesudahnya) |
| `cur.next = baru` | Menyambungkan node terakhir ke node baru |

````python
class Node:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.next = None          # pointer ke node berikutnya (awalnya kosong)


class SinglyLinkedList:
    def __init__(self):
        self.head = None              # list kosong: head menunjuk ke None

    def is_empty(self):
        return self.head is None      # True jika belum ada node

    def display(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head
        hasil = []
        while cur is not None:        # telusuri dari head sampai None
            hasil.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(hasil) + " -> None")

    def insert_front(self, data):
        baru = Node(data)             # 1. buat node baru
        baru.next = self.head         # 2. node baru menunjuk head lama
        self.head = baru              # 3. head pindah ke node baru

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():           # list kosong: node baru langsung jadi head
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:   # jalan sampai node TERAKHIR
            cur = cur.next
        cur.next = baru               # sambungkan node terakhir ke node baru


sll = SinglyLinkedList()
sll.insert_last(10)
sll.insert_last(20)
sll.insert_last(30)
sll.display()        # Output: 10 -> 20 -> 30 -> None

sll.insert_front(5)
sll.display()        # Output: 5 -> 10 -> 20 -> 30 -> None
````

## 🔢 7. Menghitung Jumlah Node (`length`)

### Konsep

Menghitung berapa banyak node dengan menelusuri list dan menambah penghitung setiap kali berpindah node.

| Baris | Penjelasan |
|---|---|
| `n = 0` | Penghitung awal |
| `n += 1` | Tambah 1 untuk setiap node yang dilewati |
| `return n` | Mengembalikan total node |

````python
class Node:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.next = None          # pointer ke node berikutnya (awalnya kosong)


class SinglyLinkedList:
    def __init__(self):
        self.head = None              # list kosong: head menunjuk ke None

    def is_empty(self):
        return self.head is None      # True jika belum ada node

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():           # list kosong: node baru langsung jadi head
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:   # jalan sampai node TERAKHIR
            cur = cur.next
        cur.next = baru               # sambungkan node terakhir ke node baru

    def length(self):
        n = 0
        cur = self.head
        while cur is not None:        # hitung setiap node yang dilewati
            n += 1
            cur = cur.next
        return n


sll = SinglyLinkedList()
print("Panjang:", sll.length())    # Output: Panjang: 0
sll.insert_last(10)
sll.insert_last(20)
sll.insert_last(30)
print("Panjang:", sll.length())    # Output: Panjang: 3
````

## 🔍 8. Mencari Data (`search`)

### Konsep

Mencari posisi (index) sebuah nilai dengan menelusuri dari `head`. Jika ketemu kembalikan index-nya (mulai dari 0), jika sampai ujung tidak ketemu kembalikan `-1`.

| Baris | Penjelasan |
|---|---|
| `index = 0` | Posisi node yang sedang dilihat |
| `if cur.data == data:` | Jika sama, kembalikan `index` |
| `return -1` | Tanda bahwa nilai tidak ditemukan |

````python
class Node:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.next = None          # pointer ke node berikutnya (awalnya kosong)


class SinglyLinkedList:
    def __init__(self):
        self.head = None              # list kosong: head menunjuk ke None

    def is_empty(self):
        return self.head is None      # True jika belum ada node

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():           # list kosong: node baru langsung jadi head
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:   # jalan sampai node TERAKHIR
            cur = cur.next
        cur.next = baru               # sambungkan node terakhir ke node baru

    def search(self, data):
        cur = self.head
        index = 0
        while cur is not None:
            if cur.data == data:      # ketemu: kembalikan posisinya
                return index
            cur = cur.next
            index += 1
        return -1                     # sampai ujung tapi tidak ketemu


sll = SinglyLinkedList()
for x in [10, 20, 30]:
    sll.insert_last(x)

print("Index 20:", sll.search(20))    # Output: Index 20: 1
print("Index 99:", sll.search(99))    # Output: Index 99: -1
````

## 🔑 Operasi Penting yang Wajib Dipahami

| Operasi | Cara Kerja | Kompleksitas |
|---|---|---|
| `is_empty()` | `head is None` | O(1) |
| `insert_front(data)` | Node baru menunjuk head lama, head pindah | O(1) |
| `insert_last(data)` | Telusuri sampai node terakhir, sambungkan | O(n) |
| `display()` | Telusuri semua node dari head | O(n) |
| `length()` | Telusuri dan hitung node | O(n) |
| `search(data)` | Telusuri sampai ketemu | O(n) |

> 🧠 **Tips kunci:** Pada `insert_front`, sambungkan `baru.next = self.head` **sebelum** memindahkan `head`. Kalau terbalik, rantai ke node lama terputus dan datanya hilang. Selalu tangani dua kasus khusus: **list kosong** dan **list dengan satu node**.
