```python
def hitung_rata(nilai, n):
    total = 0
    for i in range(n):
        total += nilai[i]
    return total / n 

def cari_nilai_tertinggi(nilai, n):
    tertinggi = nilai[0]
    i = 1
    while i < n:
        if nilai[i] > tertinggi:
            tertinggi = nilai[i]
        i += 1
    return tertinggi

def cari_nilai_terendah(nilai, n):
    terendah = nilai[0]
    i = 1
    while i < n:
        if nilai[i] < terendah:
            terendah = nilai[i]
        i += 1
    return terendah

def cari_indeks_tertinggi(nilai, n):
    index = 0
    i = 1
    while i < n:
        if nilai[i] > nilai[index]:
            index = i
        i += 1
    return index

def cari_indeks_terendah(nilai, n):
    index = 0
    i = 1
    while i < n:
        if nilai[i] < nilai[index]:
            index = i
        i += 1
    return index

def tampilkan_laporan(nama, nilai, n):
    nilai_tertinggi = cari_nilai_tertinggi(nilai, n)
    nilai_terendah = cari_nilai_terendah(nilai, n)
    index_tertinggi = cari_indeks_tertinggi(nilai, n)
    index_terendah = cari_indeks_terendah(nilai, n)
    rata_rata = hitung_rata(nilai, n)
    
    print("====================================")
    print("        LAPORAN NILAI               ")
    print("====================================")
    
    for i in range(n):
        if nilai[i] >= 60:
            status = "lulus"
        else:
            status = "tidak lulus"
        print(i + 1, nama[i], nilai[i], status)
        
    print(f"Nilai Tertinggi: {nilai_tertinggi}", nama[index_tertinggi])
    print(f"Nilai Tertinggi: {nilai_terendah}", nama[index_terendah])
    print(f"rata - rata: ", rata_rata)

nama = []
nilai = []

n = int(input("Masukkan jumlah siswa (2-10): "))

while n < 2 or n > 10:
    print("input harus dari 2 - 10")
    n = int(input("Masukkan jumlah siswa (2-10): "))
    
for i in range(n):
    print("siswa ke ", i + 1)
    nama.append((input("masukin nama: ")))
    nilai_siswa = int(input("masukin nilai: "))
    
    while nilai_siswa < 0 or nilai_siswa > 100:
        print("nilai cuma 0 - 100")
        nilai_siswa = int(input("nilai: "))
        
    nilai.append(nilai_siswa)
        
tampilkan_laporan(nama, nilai, n)
```

```python
nama_buah = ["Apel", "Mangga", "Jeruk", "Pisang", "Anggur"]
harga = [15000, 12000, 10000, 8000, 35000]
N_BUAH = 5


def hitung_subtotal(harga_satuan, berat):
    return harga_satuan * berat

def hitung_total(subtotal, n):
    total = 0

    for i in range(n):
        total = total + subtotal[i]
    return total

def hitung_kembalian(bayar, total):
    return bayar - total

def tampilkan_menu(nama_buah, harga, n):
    print("===========================")
    print("     TOKO BUAH SEGAR")
    print("===========================")

    for i in range(n):
        print(i + 1, ".", nama_buah[i], "Rp", harga[i], "/kg")
    print("")

def tampilkan_struk(nama_buah, harga, berat_beli,
                    subtotal_beli, n_beli, total, bayar):

    print("==============================")
    print("        STRUK BELANJA")
    print("==============================")

    for i in range(n_beli):
        print(i + 1, ".", nama_buah[i], berat_beli[i], "kg = Rp", subtotal_beli[i])

    print("------------------------------")
    print("Total Belanja :", total)
    print("Uang Bayar    :", bayar)
    print("Kembalian     :", hitung_kembalian(bayar, total))
    print("==============================")
    print("Terima kasih sudah berbelanja!")

buah_dibeli = []
berat_dibeli = []
subtotal_dibeli = []
n_transaksi = 0

tampilkan_menu(nama_buah, harga, N_BUAH)

while True:
    pilihan = int(input("Pilih buah (1-5): "))

    while pilihan < 1 and pilihan > 5:
        print("Pilihan harus 1-5: ")
        pilihan = int(input("Pilih buah (1-5): "))

    indeks = pilihan - 1
    berat = float(input("Berat (kg): "))

    while berat <= 0:
        print("Berat harus lebih dari 0. Coba lagi: ")
        berat = float(input("Berat (kg): "))

    subtotal = hitung_subtotal(harga[indeks], berat)

    print(nama_buah[indeks], berat, "kg = Rp", subtotal)

    buah_dibeli = buah_dibeli + [indeks]
    berat_dibeli = berat_dibeli + [berat]
    subtotal_dibeli = subtotal_dibeli + [subtotal]
    n_transaksi = n_transaksi + 1

    lagi = input("Beli buah lagi? (y/t): ")

    if lagi == "t":
        break

total = hitung_total(subtotal_dibeli, n_transaksi)
bayar = float(input("uang bayar: "))

while bayar < total:
    bayar = float(input("Uang bayar: "))

nama_struk = []

for i in range(n_transaksi):
    nama_struk = nama_struk + [nama_buah[buah_dibeli[i]]]

tampilkan_struk(
    nama_struk,
    harga,
    berat_dibeli,
    subtotal_dibeli,
    n_transaksi,
    total,
    bayar
)
```
