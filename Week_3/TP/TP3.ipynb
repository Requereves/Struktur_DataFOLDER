## Soal 1 — ADT Buku

Buatlah class `Buku` yang merepresentasikan sebuah buku di perpustakaan dengan menerapkan konsep ADT.

**Spesifikasi:**
- Atribut private: `judul` (str), `pengarang` (str), `harga` (float), `tersedia` (bool)
- Lengkapi semua operasi yang sudah disediakan di skeleton

**Operasi yang harus diimplementasikan:**

| Operasi | Keterangan |
|---|---|
| `__init__` | Konstruktor, inisialisasi semua atribut |
| `get_data()` | Kembalikan semua data dalam bentuk dictionary |
| `set_harga(harga_baru)` | Ubah harga buku |
| `is_tersedia()` | Kembalikan True jika buku tersedia |
| `pinjam()` | Set buku menjadi tidak tersedia, gagal jika sudah dipinjam |
| `kembalikan()` | Set buku kembali tersedia |
| `tampilkan_data()` | Cetak semua informasi buku |

**Petunjuk pengerjaan:** Ganti setiap `pass` dengan kode yang sesuai. Jangan mengubah nama method maupun struktur program utama. Jalankan program dan pastikan output sesuai komentar `Expected`.

```python
class Buku:
    def __init__(self, judul: str, pengarang: str, harga: float):
        self.__judul = judul
        self.__pengarang = pengarang
        self.__harga = harga
        self.__tersedia = True
        

    def get_data(self) -> dict:
        return {
            "judul": self.__judul,
            "pengarang": self.__pengarang,
            "harga": self.__harga,
            "tersedia": self.__tersedia
        }
        
    def set_harga(self, harga_baru: float):
        self.__harga = harga_baru
        
    def is_tersedia(self) -> bool:
        return self.__tersedia
        
    def pinjam(self) -> bool:
        if self.__tersedia:
            self.__tersedia = False
            print(f"Buku '{self.__judul}' berhasil dipinjam.")
            return True
        print(f"Gagal: buku '{self.__judul}' sedang dipinjam.")
        return False
        
    def kembalikan(self):
        self.__tersedia = True
        print(f"Buku '{self.__judul}' sudah dikembalikan.")

    def tampilkan_data(self):
        status = "Tersedia" if self.__tersedia else "Dipinjam"
        print(f"Judul     : {self.__judul}")
        print(f"Pengarang : {self.__pengarang}")
        print(f"Harga     : Rp{self.__harga:,.0f}")
        print(f"Status    : {status}")


# ===================== PROGRAM UTAMA =====================
if __name__ == "__main__":
    # 1. Buat objek Buku
    buku1 = Buku("Laskar Pelangi", "Andrea Hirata", 85000)

    # 2. Tampilkan data awal
    print("=== Data Buku ===")
    buku1.tampilkan_data()

    # 3. Pinjam buku
    print("\n--- Meminjam buku ---")
    buku1.pinjam()
    print("Tersedia:", buku1.is_tersedia())  # Expected: False

    # 4. Coba pinjam lagi (seharusnya gagal)
    print("\n--- Meminjam lagi ---")
    buku1.pinjam()  # Expected: pesan gagal

    # 5. Kembalikan buku
    print("\n--- Mengembalikan buku ---")
    buku1.kembalikan()
    print("Tersedia:", buku1.is_tersedia())  # Expected: True

    # 6. Ubah harga dan tampilkan dalam bentuk dictionary
    buku1.set_harga(90000)
    print("\nData Dictionary:", buku1.get_data())
```
