## Apa Itu Queue?

**Queue** adalah struktur data yang mengikuti prinsip **FIFO (First In, First Out)** — elemen yang **pertama dimasukkan** adalah yang **pertama dikeluarkan**.

Bayangkan seperti antrian orang: orang yang pertama mengantri adalah yang pertama dilayani.

```
FRONT -->  [  10  ]  [  20  ]  [  30  ]  <-- REAR
           --------  --------  --------
           (pertama masuk,              (terakhir masuk,
            pertama keluar)              terakhir keluar)
```

Setiap **queue** terdiri dari:
- `data` → nilai yang disimpan di setiap elemen
- `front` → pointer/indeks yang menunjuk ke elemen **paling depan** (akan keluar berikutnya)
- `rear` → pointer/indeks yang menunjuk ke elemen **paling belakang** (terakhir masuk)

**Operasi utama Queue:**
- `enqueue` → menambahkan elemen ke belakang antrian
- `dequeue` → mengambil dan menghapus elemen terdepan
- `peek` → melihat elemen terdepan tanpa menghapusnya
- `is_empty` → mengecek apakah queue kosong

**`front`** selalu menunjuk ke elemen yang akan keluar berikutnya.  
**`rear`** selalu menunjuk ke elemen yang terakhir masuk.  
Jika queue kosong, biasanya `front = rear = -1` (implementasi array) atau `front = rear = None` (implementasi linked list).


## 📦 1. Membuat Queue dan Menginisialisasi

### Konsep
Sebelum bisa melakukan apa pun, kita perlu membuat "cetakan" queue menggunakan `class`. Queue bisa diimplementasikan dengan **list Python** (paling sederhana) atau **linked list**. Di sini kita gunakan list Python.

| Baris | Penjelasan |
|---|---|
| `class Queue:` | Membuat cetak biru (blueprint) untuk queue |
| `def __init__(self):` | Fungsi yang otomatis dipanggil saat `Queue()` dibuat |
| `self.items = []` | List kosong sebagai wadah — elemen queue disimpan di sini |

````python
# Definisi kelas Queue
class Queue:
    def __init__(self):         # Konstruktor: dipanggil saat Queue baru dibuat
        self.items = []         # List kosong sebagai wadah elemen queue

q = Queue()
print(q.items)   # Output: []
````

## ✅ 2. Memeriksa Apakah Queue Kosong (`is_empty`)

### Konsep
Sebelum melakukan operasi seperti `dequeue` atau `peek`, kita perlu tahu apakah queue punya isi atau tidak. Caranya sederhana: cek apakah `items` bernilai kosong.

````python
class Queue:
    def __init__(self):
        self.items = []
    
    def is_empty(self):
        return len(self.items) == 0   # Mengembalikan True jika kosong, False jika tidak

q = Queue()
print(q.is_empty())   # Output: True
````

## ➕ 3. Menambahkan Elemen (`enqueue`)

### Konsep
Menambahkan elemen **baru** ke posisi **paling belakang** queue. Elemen yang baru ditambahkan langsung menjadi `rear`.

Langkah-langkahnya:
1. Ambil elemen baru
2. Letakkan di belakang antrian (akhir list)
3. Elemen baru menjadi rear

```
Sebelum enqueue(30):                      Sesudah enqueue(30):
FRONT --> [ 10 ][ 20 ] <-- REAR    FRONT --> [ 10 ][ 20 ][ 30 ] <-- REAR (baru)
```

| Baris | Penjelasan |
|---|---|
| `self.items.append(data)` | Menambahkan elemen baru ke bagian belakang (akhir list = rear queue) |

````python
class Queue:
    def __init__(self):
        self.items = []
    
    def enqueue(self, data):
        self.items.append(data)      # Tambahkan elemen baru ke belakang queue (akhir list)
        print(f"Enqueue {data} ke queue")

q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
print(q.items)   # Output: [10, 20, 30]  (10 adalah front, 30 adalah rear)
````

## ❌ 4. Mengambil Elemen Terdepan (`dequeue`)

### Konsep
Mengambil dan **menghapus** elemen yang berada di **paling depan** queue (prinsip FIFO). Elemen yang pertama masuk adalah yang pertama keluar.

```
Sebelum dequeue():                          Sesudah dequeue():
FRONT --> [ 10 ][ 20 ][ 30 ] <-- REAR    FRONT --> [ 20 ][ 30 ] <-- REAR
          (10 diambil dan dihapus)
```

| Baris | Penjelasan |
|---|---|
| `if self.is_empty():` | Selalu cek kosong sebelum dequeue untuk menghindari error |
| `self.items.pop(0)` | Mengambil dan menghapus elemen pertama (front queue) |
| `return dequeued` | Mengembalikan nilai yang diambil agar bisa dipakai |

````python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, data):
        self.items.append(data)

    def dequeue(self):
        if self.is_empty():              # Cek dulu apakah queue kosong
            print("Queue kosong, tidak ada yang bisa di-dequeue!")
            return None
        dequeued = self.items.pop(0)     # Ambil dan hapus elemen terdepan (FIFO)
        print(f"Dequeue: {dequeued} dikeluarkan dari queue")
        return dequeued                  # Kembalikan nilai yang diambil

    def is_empty(self):
        return len(self.items) == 0

q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.dequeue()        # Output: Dequeue: 10 dikeluarkan dari queue
q.dequeue()        # Output: Dequeue: 20 dikeluarkan dari queue
print(q.items)     # Output: [30]
````

## 🔍 5. Melihat Elemen Terdepan (`peek`)

### Konsep
Melihat nilai elemen **paling depan** queue **tanpa menghapusnya**. Berbeda dengan `dequeue`, elemen tetap ada di queue setelah `peek`.

```
Queue saat ini:                              peek() mengembalikan 10, queue TIDAK berubah:
FRONT --> [ 10 ][ 20 ][ 30 ] <-- REAR    FRONT --> [ 10 ][ 20 ][ 30 ] <-- REAR
          (10 adalah front,                         (queue TIDAK berubah)
           akan dikembalikan)
```

| Baris | Penjelasan |
|---|---|
| `self.items[0]` | Indeks `0` di Python berarti elemen pertama (paling depan/front queue) |
| Tidak ada `.pop(0)` | Hanya membaca nilai, tidak menghapus |

````python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, data):
        self.items.append(data)

    def is_empty(self):
        return len(self.items) == 0

    def peek(self):
        if self.is_empty():              # Cek apakah queue kosong
            print("Queue kosong!")
            return None
        return self.items[0]             # Kembalikan elemen terdepan TANPA menghapusnya

q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
print(q.peek())    # Output: 10
print(q.items)     # Output: [10, 20, 30]  (tidak berubah)
````

## 🔢 6. Menghitung Jumlah Elemen (`size`)

### Konsep
Menghitung berapa banyak elemen yang ada di dalam queue saat ini.

````python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, data):
        self.items.append(data)

    def size(self):
        return len(self.items)           # Kembalikan jumlah elemen dalam queue

q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
print("Jumlah elemen:", q.size())   # Output: Jumlah elemen: 3
````

## 🖥️ 7. Menampilkan Isi Queue (`display`)

### Konsep
Menampilkan seluruh elemen queue dari **depan (front) ke belakang (rear)**, sehingga terlihat seperti antrian yang sesungguhnya.

| Baris | Penjelasan |
|---|---|
| `self.items` | Urutan list sudah sesuai — front ada di index 0 |
| `for item in ...` | Kunjungi setiap elemen dari depan ke belakang |

````python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, data):
        self.items.append(data)

    def is_empty(self):
        return len(self.items) == 0

    def display(self):
        if self.is_empty():              # Cek apakah queue kosong
            print("Queue kosong")
            return

        print("Queue (dari depan ke belakang):")
        print("FRONT --> ", end="")
        for item in self.items:          # Iterasi dari depan ke belakang (front ke rear)
            print(f"[ {item} ]", end=" ")
        print("<-- REAR")

q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.display()
# Queue (dari depan ke belakang):
# FRONT --> [ 10 ] [ 20 ] [ 30 ] <-- REAR
````

## 🧹 8. Mengosongkan Queue (`clear`)

### Konsep
Menghapus semua elemen dalam queue sekaligus hingga menjadi kosong.

````python
class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, data):
        self.items.append(data)

    def display(self):
        if len(self.items) == 0:
            print("Queue kosong")
            return
        print("Queue (dari depan ke belakang):")
        print("FRONT --> ", end="")
        for item in self.items:
            print(f"[ {item} ]", end=" ")
        print("<-- REAR")

    def clear(self):
        self.items = []                  # Reset list menjadi kosong
        print("Queue berhasil dikosongkan")

q = Queue()
q.enqueue(10)
q.enqueue(20)
q.enqueue(30)
q.display()    # Ada 3 elemen
q.clear()      # Output: Queue berhasil dikosongkan
q.display()    # Output: Queue kosong
````

````python
# ===================== DEMO =====================
class Queue:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def enqueue(self, data):
        self.items.append(data)
        print(f"Enqueue {data} ke queue")

    def dequeue(self):
        if self.is_empty():
            print("Queue kosong, tidak ada yang bisa di-dequeue!")
            return None
        dequeued = self.items.pop(0)
        print(f"Dequeue: {dequeued} dikeluarkan dari queue")
        return dequeued

    def peek(self):
        if self.is_empty():
            print("Queue kosong!")
            return None
        return self.items[0]

    def size(self):
        return len(self.items)

    def display(self):
        if self.is_empty():
            print("Queue kosong")
            return
        print("Queue (dari depan ke belakang):")
        print("FRONT --> ", end="")
        for item in self.items:
            print(f"[ {item} ]", end=" ")
        print("<-- REAR")

    def clear(self):
        self.items = []
        print("Queue berhasil dikosongkan")


if __name__ == "__main__":
    q = Queue()

    print("=== Enqueue ===")
    q.enqueue(10)
    q.enqueue(20)
    q.enqueue(30)
    q.enqueue(40)
    q.display()
    print("Jumlah elemen:", q.size())    # 4

    print("\n=== Peek ===")
    print("Front saat ini:", q.peek())   # 10

    print("\n=== Dequeue ===")
    q.dequeue()
    q.dequeue()
    q.display()

    print("\n=== Clear ===")
    q.clear()
    q.display()
````

## 📐 Konsep Queue: Ringkasan Visual

```
KONDISI AWAL (queue kosong):
items = []

SETELAH enqueue(10):
items = [10]
         ^
        FRONT & REAR

SETELAH enqueue(20):
items = [10, 20]
         ^       ^
        FRONT   REAR

SETELAH enqueue(30):
items = [10, 20, 30]
         ^           ^
        FRONT       REAR

SAAT dequeue():
  Ambil items[0] = 10  →  kembalikan 10
  items = [20, 30]
           ^       ^
          FRONT   REAR baru

SAAT peek():
  Baca items[0] = 20  →  kembalikan 20 (items TIDAK berubah)
  items = [20, 30]
```


## 🔑 Operasi Penting yang Wajib Dipahami

| Operasi | Cara Kerja | Kompleksitas |
|---|---|---|
| `enqueue(data)` | `items.append(data)` — tambah ke akhir list | O(1) |
| `dequeue()` | `items.pop(0)` — hapus dari awal list | O(n) |
| `peek()` | `items[0]` — baca awal list tanpa hapus | O(1) |
| `is_empty()` | `len(items) == 0` — cek kosong | O(1) |
| `size()` | `len(items)` — hitung elemen | O(1) |
| `clear()` | `items = []` — reset queue | O(1) |

> 🧠 **Tips kunci:** Queue bekerja dengan prinsip FIFO — `enqueue` menambahkan di **belakang (REAR)**, sedangkan `dequeue` mengambil dari **depan (FRONT)**. Gunakan `peek()` jika hanya ingin melihat elemen terdepan tanpa mengubah isi queue. Selalu cek `is_empty()` sebelum `dequeue()` atau `peek()` untuk menghindari error!
