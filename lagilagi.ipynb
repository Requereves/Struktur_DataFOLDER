## Pekan 1 - Pointer dan ADT

**Soal:**

Buat ADT **Titik** (koordinat x, y) dengan operasi `geser(dx, dy)`, `jarak(lain)` (jarak Euclid), dan `tampil()`.  
(a) Hitung jarak A(0, 0) ke B(3, 4).  
(b) Geser A sebesar (1, 1).  
(c) Buat salinan `d = Titik(b.x, b.y)`, lalu alias `c = b`, lalu geser lewat `c`. Tunjukkan mana yang ikut berubah dan mana yang tidak (alias vs salinan).

````python
# Jawaban
from math import sqrt


class Titik:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def geser(self, dx, dy):
        self.x += dx
        self.y += dy

    def jarak(self, lain):
        return sqrt((self.x - lain.x) ** 2 + (self.y - lain.y) ** 2)

    def tampil(self):
        return f"({self.x}, {self.y})"


def soal1():
    a = Titik(0, 0)
    b = Titik(3, 4)
    print("(a) Jarak A-B =", a.jarak(b))
    a.geser(1, 1)
    print("(b) A setelah geser =", a.tampil())
    d = Titik(b.x, b.y)    # salinan: objek BARU
    c = b                  # alias: menunjuk objek yang SAMA
    c.geser(1, 1)
    print("(c) B setelah c.geser(1, 1) =", b.tampil())
    print("(c) salinan D tetap =", d.tampil())
    print("(c) c is b?", c is b, "| d is b?", d is b)
````

````python
soal1()
````

**Output yang diharapkan:**

```
(a) Jarak A-B = 5.0
(b) A setelah geser = (1, 1)
(c) B setelah c.geser(1, 1) = (4, 5)
(c) salinan D tetap = (3, 4)
(c) c is b? True | d is b? False
```

## Pekan 2 - Singly Linked List (playlist)

**Soal:**

Buat Singly Linked List untuk playlist lagu (data berupa string) dengan operasi `insert_front`, `insert_last`, `hapus_depan`, `hapus_belakang`, `sisip_setelah(target, data)`, dan `panjang()`.  
Masukkan Jazz, Pop, Rock di belakang, lalu Folk di depan. Hapus depan, hapus belakang, lalu sisipkan Blues setelah Jazz.  
Tampilkan isi list di tiap tahap dan panjang akhirnya.

````python
# Jawaban
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def insert_front(self, data):
        baru = Node(data)
        baru.next = self.head
        self.head = baru

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:
            cur = cur.next
        cur.next = baru

    def hapus_depan(self):
        if self.is_empty():
            print("List kosong")
            return
        self.head = self.head.next

    def hapus_belakang(self):
        if self.is_empty():
            print("List kosong")
            return
        if self.head.next is None:      # hanya 1 node
            self.head = None
            return
        cur = self.head
        while cur.next.next is not None:  # berhenti di node sebelum terakhir
            cur = cur.next
        cur.next = None

    def sisip_setelah(self, target, data):
        cur = self.head
        while cur is not None and cur.data != target:
            cur = cur.next
        if cur is None:
            print(f"{target} tidak ditemukan")
            return
        baru = Node(data)
        baru.next = cur.next            # sambung ke depan dulu
        cur.next = baru

    def panjang(self):
        n, cur = 0, self.head
        while cur is not None:
            n += 1
            cur = cur.next
        return n

    def isi(self):
        cur, hasil = self.head, []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next
        return " -> ".join(hasil + ["None"])


def tampil_list(label, lst):
    print(f"{label:<16}: {lst.isi()}")


def soal2():
    pl = SinglyLinkedList()
    for lagu in ["Jazz", "Pop", "Rock"]:
        pl.insert_last(lagu)
    pl.insert_front("Folk")
    tampil_list("Awal", pl)
    pl.hapus_depan()
    tampil_list("Hapus depan", pl)
    pl.hapus_belakang()
    tampil_list("Hapus belakang", pl)
    pl.sisip_setelah("Jazz", "Blues")
    tampil_list("Sisip Blues", pl)
    print("Panjang akhir:", pl.panjang())
````

````python
soal2()
````

**Output yang diharapkan:**

```
Awal            : Folk -> Jazz -> Pop -> Rock -> None
Hapus depan     : Jazz -> Pop -> Rock -> None
Hapus belakang  : Jazz -> Pop -> None
Sisip Blues     : Jazz -> Blues -> Pop -> None
Panjang akhir: 3
```

## Pekan 3 - Singly Linked List (hapus index dan gabung)

**Soal:**

Gunakan kelas Singly Linked List dari Pekan 2.  
(a) Buat method `hapus_index(k)` yang menghapus node pada index ke-k (index mulai dari 0). Jika index tidak valid, cetak "Index tidak valid". Uji pada list 10, 20, 30, 40: hapus index 2, hapus index 0, lalu hapus index 9.  
(b) Buat fungsi `gabung_terurut(a, b)` yang menggabungkan dua list yang sudah terurut menaik menjadi satu list baru yang tetap terurut. Uji dengan A = 1, 4, 7 dan B = 2, 3, 9.

````python
# Jawaban
class SLLLanjut(SinglyLinkedList):
    def hapus_index(self, k):
        if self.is_empty() or k < 0:
            print("Index tidak valid")
            return
        if k == 0:
            self.head = self.head.next
            return
        cur, i = self.head, 0
        while cur is not None and i < k - 1:   # berhenti di node SEBELUM index k
            cur = cur.next
            i += 1
        if cur is None or cur.next is None:
            print("Index tidak valid")
            return
        cur.next = cur.next.next


def gabung_terurut(a, b):
    hasil = SinglyLinkedList()
    p, q = a.head, b.head
    while p is not None and q is not None:
        if p.data <= q.data:
            hasil.insert_last(p.data)
            p = p.next
        else:
            hasil.insert_last(q.data)
            q = q.next
    while p is not None:                # sisa list A
        hasil.insert_last(p.data)
        p = p.next
    while q is not None:                # sisa list B
        hasil.insert_last(q.data)
        q = q.next
    return hasil


def soal3():
    sll = SLLLanjut()
    for x in [10, 20, 30, 40]:
        sll.insert_last(x)
    tampil_list("Awal", sll)
    sll.hapus_index(2)
    tampil_list("Hapus index 2", sll)
    sll.hapus_index(0)
    tampil_list("Hapus index 0", sll)
    sll.hapus_index(9)
    a, b = SinglyLinkedList(), SinglyLinkedList()
    for x in [1, 4, 7]:
        a.insert_last(x)
    for x in [2, 3, 9]:
        b.insert_last(x)
    tampil_list("List A", a)
    tampil_list("List B", b)
    tampil_list("Gabungan", gabung_terurut(a, b))
````

````python
soal3()
````

**Output yang diharapkan:**

```

Hapus index 2   : 10 -> 20 -> 40 -> None
Hapus index 0   : 20 -> 40 -> None
Index tidak valid
List A          : 1 -> 4 -> 7 -> None
List B          : 2 -> 3 -> 9 -> None
Gabungan        : 1 -> 2 -> 3 -> 4 -> 7 -> 9 -> None
```

## Pekan 4 - Doubly Linked List (riwayat browser)

**Soal:**

Buat riwayat browser dengan Doubly Linked List. Operasi: `kunjungi(halaman)` (tambah halaman baru setelah halaman sekarang dan buang riwayat "maju" yang lama), `mundur()`, dan `maju()`. Jika tidak bisa mundur/maju, cetak pesan.  
Skenario: mulai di google.com, kunjungi youtube.com, kunjungi github.com, mundur 3 kali, maju 1 kali, kunjungi wikipedia.org, lalu maju.

````python
# Jawaban
class DNode:
    def __init__(self, data):
        self.data = data
        self.prev = None
        self.next = None


class Browser:
    def __init__(self, beranda):
        self.cur = DNode(beranda)       # cur = halaman yang sedang dibuka

    def kunjungi(self, halaman):
        baru = DNode(halaman)
        self.cur.next = baru            # riwayat 'maju' lama otomatis terbuang
        baru.prev = self.cur
        self.cur = baru

    def mundur(self):
        if self.cur.prev is None:
            print("Tidak bisa mundur")
        else:
            self.cur = self.cur.prev

    def maju(self):
        if self.cur.next is None:
            print("Tidak bisa maju")
        else:
            self.cur = self.cur.next


def soal4():
    b = Browser("google.com")
    b.kunjungi("youtube.com")
    b.kunjungi("github.com")
    print("Sekarang di :", b.cur.data)
    b.mundur()
    print("Mundur      :", b.cur.data)
    b.mundur()
    print("Mundur      :", b.cur.data)
    b.mundur()
    print("Mundur      :", b.cur.data)
    b.maju()
    print("Maju        :", b.cur.data)
    b.kunjungi("wikipedia.org")
    print("Kunjungi    :", b.cur.data)
    b.maju()
    print("Maju        :", b.cur.data)
````

```
Tidak bisa mundur
Mundur      : google.com

Tidak bisa maju
Maju        : wikipedia.org
```

## Pekan 5 - Stack (undo dan redo)

**Soal:**

Buat kelas Stack (`push`, `pop`, `peek`, `is_empty`). Lalu buat editor teks sederhana dengan dua Stack: `tulis(kata)`, `undo()`, `redo()`, dan `isi()`.  
Aturan: `undo` memindahkan kata terakhir ke stack redo, `redo` mengembalikannya, dan menulis kata baru mengosongkan stack redo.  
Skenario: tulis Halo, Dunia, !, lalu undo 2 kali, redo 1 kali, tulis Python, lalu redo.

````python
# Jawaban
class Stack:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def push(self, data):
        self.items.append(data)

    def pop(self):
        if self.is_empty():
            return None
        return self.items.pop()

    def peek(self):
        return None if self.is_empty() else self.items[-1]


class Editor:
    def __init__(self):
        self.undo_stack = Stack()       # kata-kata yang sedang tertulis
        self.redo_stack = Stack()       # kata-kata yang di-undo

    def tulis(self, kata):
        self.undo_stack.push(kata)
        self.redo_stack = Stack()       # tulis baru = riwayat redo hangus

    def undo(self):
        if self.undo_stack.is_empty():
            print("Tidak ada yang bisa di-undo")
            return
        self.redo_stack.push(self.undo_stack.pop())

    def redo(self):
        if self.redo_stack.is_empty():
            print("Tidak ada yang bisa di-redo")
            return
        self.undo_stack.push(self.redo_stack.pop())

    def isi(self):
        return " ".join(self.undo_stack.items)


def soal5():
    e = Editor()
    for kata in ["Halo", "Dunia", "!"]:
        e.tulis(kata)
    print("Tulis x3 :", e.isi())
    e.undo()
    print("Undo     :", e.isi())
    e.undo()
    print("Undo     :", e.isi())
    e.redo()
    print("Redo     :", e.isi())
    e.tulis("Python")
    print("Tulis    :", e.isi())
    e.redo()
    print("Redo     :", e.isi())
````

````python
soal5()
````

**Output yang diharapkan:**

```
Tulis x3 : Halo Dunia !
Undo     : Halo Dunia
Undo     : Halo
Redo     : Halo Dunia
Tulis    : Halo Dunia Python
Tidak ada yang bisa di-redo
Redo     : Halo Dunia Python
```

## Pekan 6 - Queue (Round Robin)

**Soal:**

Simulasikan penjadwalan CPU **Round Robin** dengan Queue. Tiap proses punya (nama, waktu_butuh). Setiap giliran, proses berjalan maksimal `quantum` detik. Jika belum selesai, proses masuk lagi ke belakang antrian.  
Data: P1 (5 detik), P2 (3 detik), P3 (8 detik), quantum = 3.  
Cetak waktu saat giliran selesai, lama jalan, dan sisa waktu tiap giliran.

````python
# Jawaban
class Queue:
    def __init__(self):
        self.items = []

    def is_empty(self):
        return len(self.items) == 0

    def enqueue(self, data):
        self.items.append(data)

    def dequeue(self):
        if self.is_empty():
            return None
        return self.items.pop(0)


def round_robin(proses, quantum):
    q = Queue()
    for nama, butuh in proses:
        q.enqueue([nama, butuh])
    waktu = 0
    while not q.is_empty():
        nama, sisa = q.dequeue()
        jalan = min(quantum, sisa)
        waktu += jalan
        sisa -= jalan
        info = f"t={waktu:>2}: {nama} jalan {jalan} detik, sisa {sisa}"
        if sisa == 0:
            print(info + " -> SELESAI")
        else:
            print(info)
            q.enqueue([nama, sisa])     # kembali ke belakang antrian


def soal6():
    round_robin([("P1", 5), ("P2", 3), ("P3", 8)], quantum=3)
````

````python
soal6()
````

**Output yang diharapkan:**

```
t= 3: P1 jalan 3 detik, sisa 2
t= 6: P2 jalan 3 detik, sisa 0 -> SELESAI
t= 9: P3 jalan 3 detik, sisa 5
t=11: P1 jalan 2 detik, sisa 0 -> SELESAI
t=14: P3 jalan 3 detik, sisa 2
t=16: P3 jalan 2 detik, sisa 0 -> SELESAI
```
