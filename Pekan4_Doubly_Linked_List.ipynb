# Pekan 4 - Doubly Linked List

## Apa Itu Doubly Linked List?

**Doubly Linked List** adalah linked list di mana setiap node punya **dua pointer**: `next` (ke node berikutnya) dan `prev` (ke node sebelumnya). Karena itu list bisa ditelusuri **maju maupun mundur**.

```
        head                                    tail
         |                                       |
         v                                       v
None <-[ 10 ]<->[ 20 ]<->[ 30 ]-> None
       prev next
```

Setiap node terdiri dari:
- `data` → nilai yang disimpan
- `prev` → pointer ke node sebelumnya (`None` untuk node pertama)
- `next` → pointer ke node berikutnya (`None` untuk node terakhir)

List ini menyimpan **`head`** (node pertama) dan **`tail`** (node terakhir), sehingga `insert_last` dan `delete_last` bisa dilakukan tanpa menelusuri seluruh list.

**Operasi yang dibahas:**
- `is_empty` → mengecek apakah list kosong
- `display_forward` / `display_backward` → menampilkan maju / mundur
- `insert_front`, `insert_last`, `insert_after` → menambah node
- `delete_front`, `delete_last`, `delete_value` → menghapus node

Jika list kosong, `head = tail = None`. Jika hanya satu node, `head` dan `tail` menunjuk node yang sama.

## 📦 1. Membuat Node (`DNode`)

### Konsep

Node pada doubly linked list punya **dua** pointer. Saat dibuat, keduanya `None`.

| Baris | Penjelasan |
|---|---|
| `self.data = data` | Nilai yang disimpan |
| `self.prev = None` | Pointer ke node sebelumnya |
| `self.next = None` | Pointer ke node berikutnya |

````python
class DNode:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.prev = None          # pointer ke node SEBELUMNYA
        self.next = None          # pointer ke node BERIKUTNYA


n1 = DNode(10)
n2 = DNode(20)
n1.next = n2          # n1 maju ke n2
n2.prev = n1          # n2 mundur ke n1

print(n1.next.data)   # Output: 20
print(n2.prev.data)   # Output: 10
````

## 🏗️ 2. Membuat Class `DoublyLinkedList`

### Konsep

Berbeda dengan singly linked list, kita menyimpan **dua** penunjuk: `head` (ujung depan) dan `tail` (ujung belakang).

| Baris | Penjelasan |
|---|---|
| `self.head = None` | Node pertama, kosong di awal |
| `self.tail = None` | Node terakhir, kosong di awal |

````python
class DNode:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.prev = None          # pointer ke node SEBELUMNYA
        self.next = None          # pointer ke node BERIKUTNYA


class DoublyLinkedList:
    def __init__(self):
        self.head = None              # node pertama
        self.tail = None              # node terakhir


dll = DoublyLinkedList()
print(dll.head, dll.tail)    # Output: None None
````

## ✅ 3. Memeriksa Apakah List Kosong (`is_empty`)

### Konsep

List kosong jika `head` adalah `None` (otomatis `tail` juga `None`).

| Baris | Penjelasan |
|---|---|
| `self.head is None` | `True` jika tidak ada node |

````python
class DNode:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.prev = None          # pointer ke node SEBELUMNYA
        self.next = None          # pointer ke node BERIKUTNYA


class DoublyLinkedList:
    def __init__(self):
        self.head = None              # node pertama
        self.tail = None              # node terakhir

    def is_empty(self):
        return self.head is None


dll = DoublyLinkedList()
print(dll.is_empty())        # Output: True
dll.head = dll.tail = DNode(10)
print(dll.is_empty())        # Output: False
````

## 🖥️ 4. Menampilkan Maju dan Mundur

### Konsep

Inilah keunggulan utama doubly linked list: kita bisa menelusuri dari **head lewat `next`** (maju) atau dari **tail lewat `prev`** (mundur).

| Baris | Penjelasan |
|---|---|
| `cur = self.head` + `cur.next` | Telusuri maju dari depan |
| `cur = self.tail` + `cur.prev` | Telusuri mundur dari belakang |
| `" <-> ".join(hasil)` | Tanda panah dua arah |

````python
class DNode:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.prev = None          # pointer ke node SEBELUMNYA
        self.next = None          # pointer ke node BERIKUTNYA


class DoublyLinkedList:
    def __init__(self):
        self.head = None              # node pertama
        self.tail = None              # node terakhir

    def is_empty(self):
        return self.head is None

    def display_forward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head               # mulai dari depan
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next            # maju lewat next
        print("Maju  : " + " <-> ".join(hasil))

    def display_backward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.tail               # mulai dari belakang
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.prev            # mundur lewat prev
        print("Mundur: " + " <-> ".join(hasil))


dll = DoublyLinkedList()
dll.display_forward()            # Output: List kosong

n1, n2, n3 = DNode(10), DNode(20), DNode(30)     # rantai manual
n1.next, n2.prev = n2, n1
n2.next, n3.prev = n3, n2
dll.head, dll.tail = n1, n3

dll.display_forward()            # Output: Maju  : 10 <-> 20 <-> 30
dll.display_backward()           # Output: Mundur: 30 <-> 20 <-> 10
````

## ➕ 5. Menambah di Depan (`insert_front`)

### Konsep

Dua pointer harus diatur: node baru menunjuk maju ke head lama, dan head lama menunjuk **mundur** ke node baru. Operasi O(1).

Kasus khusus: jika list kosong, `head` dan `tail` sama-sama menjadi node baru.

```
Sebelum insert_front(5):     None <-[10]<->[20]-> None

Sesudah:   None <-[5]<->[10]<->[20]-> None
                   ^
                  head baru
```

| Baris | Penjelasan |
|---|---|
| `baru.next = self.head` | Node baru menunjuk maju ke head lama |
| `self.head.prev = baru` | Head lama menunjuk **mundur** ke node baru |
| `self.head = baru` | Head pindah ke node baru |

````python
class DNode:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.prev = None          # pointer ke node SEBELUMNYA
        self.next = None          # pointer ke node BERIKUTNYA


class DoublyLinkedList:
    def __init__(self):
        self.head = None              # node pertama
        self.tail = None              # node terakhir

    def is_empty(self):
        return self.head is None

    def display_forward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head               # mulai dari depan
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next            # maju lewat next
        print("Maju  : " + " <-> ".join(hasil))

    def display_backward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.tail               # mulai dari belakang
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.prev            # mundur lewat prev
        print("Mundur: " + " <-> ".join(hasil))

    def insert_front(self, data):
        baru = DNode(data)
        if self.is_empty():           # list kosong: head dan tail sama-sama node baru
            self.head = self.tail = baru
        else:
            baru.next = self.head     # 1. baru menunjuk head lama
            self.head.prev = baru     # 2. head lama menunjuk balik ke baru
            self.head = baru          # 3. head pindah ke baru


dll = DoublyLinkedList()
dll.insert_front(30)
dll.insert_front(20)
dll.insert_front(10)
dll.display_forward()        # Output: Maju  : 10 <-> 20 <-> 30
dll.display_backward()       # Output: Mundur: 30 <-> 20 <-> 10
````

## ➕ 6. Menambah di Belakang (`insert_last`)

### Konsep

Karena ada `tail`, kita **tidak perlu menelusuri** list. Node baru menunjuk mundur ke tail lama, tail lama menunjuk maju ke node baru, lalu `tail` pindah. Operasi O(1) (di singly linked list ini O(n)).

| Baris | Penjelasan |
|---|---|
| `baru.prev = self.tail` | Node baru menunjuk mundur ke tail lama |
| `self.tail.next = baru` | Tail lama menunjuk maju ke node baru |
| `self.tail = baru` | Tail pindah ke node baru |

````python
class DNode:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.prev = None          # pointer ke node SEBELUMNYA
        self.next = None          # pointer ke node BERIKUTNYA


class DoublyLinkedList:
    def __init__(self):
        self.head = None              # node pertama
        self.tail = None              # node terakhir

    def is_empty(self):
        return self.head is None

    def display_forward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head               # mulai dari depan
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next            # maju lewat next
        print("Maju  : " + " <-> ".join(hasil))

    def display_backward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.tail               # mulai dari belakang
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.prev            # mundur lewat prev
        print("Mundur: " + " <-> ".join(hasil))

    def insert_last(self, data):
        baru = DNode(data)
        if self.is_empty():
            self.head = self.tail = baru
        else:
            baru.prev = self.tail     # 1. baru menunjuk tail lama (mundur)
            self.tail.next = baru     # 2. tail lama menunjuk baru (maju)
            self.tail = baru          # 3. tail pindah ke baru


dll = DoublyLinkedList()
dll.insert_last(10)
dll.insert_last(20)
dll.insert_last(30)
dll.display_forward()        # Output: Maju  : 10 <-> 20 <-> 30
dll.display_backward()       # Output: Mundur: 30 <-> 20 <-> 10
````

## ➕ 7. Menyisipkan Setelah Node Tertentu (`insert_after`)

### Konsep

Menyisipkan di tengah butuh **empat** pointer diatur. Jika target adalah node terakhir, cukup panggil `insert_last`.

```
insert_after(20, 25):

[10]<->[20]<->[30]        [10]<->[20]<->[25]<->[30]
        cur                       cur    baru
```

| Baris | Penjelasan |
|---|---|
| `baru.prev = cur` | Baru menunjuk mundur ke target |
| `baru.next = cur.next` | Baru menunjuk maju ke node setelah target |
| `cur.next.prev = baru` | Node setelah target menunjuk mundur ke baru |
| `cur.next = baru` | Target menunjuk maju ke baru |

````python
class DNode:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.prev = None          # pointer ke node SEBELUMNYA
        self.next = None          # pointer ke node BERIKUTNYA


class DoublyLinkedList:
    def __init__(self):
        self.head = None              # node pertama
        self.tail = None              # node terakhir

    def is_empty(self):
        return self.head is None

    def display_forward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head               # mulai dari depan
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next            # maju lewat next
        print("Maju  : " + " <-> ".join(hasil))

    def display_backward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.tail               # mulai dari belakang
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.prev            # mundur lewat prev
        print("Mundur: " + " <-> ".join(hasil))

    def insert_last(self, data):
        baru = DNode(data)
        if self.is_empty():
            self.head = self.tail = baru
        else:
            baru.prev = self.tail     # 1. baru menunjuk tail lama (mundur)
            self.tail.next = baru     # 2. tail lama menunjuk baru (maju)
            self.tail = baru          # 3. tail pindah ke baru

    def insert_after(self, target, data):
        cur = self.head
        while cur is not None and cur.data != target:
            cur = cur.next
        if cur is None:
            print(f"{target} tidak ditemukan")
            return
        if cur is self.tail:          # target adalah node terakhir
            self.insert_last(data)
            return
        baru = DNode(data)
        baru.prev = cur               # 1. baru menunjuk mundur ke target
        baru.next = cur.next          # 2. baru menunjuk maju ke node setelah target
        cur.next.prev = baru          # 3. node setelah target menunjuk mundur ke baru
        cur.next = baru               # 4. target menunjuk maju ke baru


dll = DoublyLinkedList()
for x in [10, 20, 30]:
    dll.insert_last(x)

dll.insert_after(20, 25)
dll.display_forward()        # Output: Maju  : 10 <-> 20 <-> 25 <-> 30
dll.display_backward()       # Output: Mundur: 30 <-> 25 <-> 20 <-> 10

dll.insert_after(30, 40)     # target = tail
dll.display_forward()        # Output: Maju  : 10 <-> 20 <-> 25 <-> 30 <-> 40
dll.insert_after(99, 1)      # Output: 99 tidak ditemukan
````

## ❌ 8. Menghapus Node Pertama (`delete_front`)

### Konsep

Pindahkan `head` ke node kedua, lalu set `prev` head baru menjadi `None`. Jika hanya ada satu node, `head` dan `tail` menjadi `None`.

| Baris | Penjelasan |
|---|---|
| `if self.head is self.tail:` | Hanya 1 node: list menjadi kosong |
| `self.head = self.head.next` | Head pindah ke node kedua |
| `self.head.prev = None` | Memutus pointer mundur ke node yang dihapus |

````python
class DNode:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.prev = None          # pointer ke node SEBELUMNYA
        self.next = None          # pointer ke node BERIKUTNYA


class DoublyLinkedList:
    def __init__(self):
        self.head = None              # node pertama
        self.tail = None              # node terakhir

    def is_empty(self):
        return self.head is None

    def display_forward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head               # mulai dari depan
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next            # maju lewat next
        print("Maju  : " + " <-> ".join(hasil))

    def display_backward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.tail               # mulai dari belakang
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.prev            # mundur lewat prev
        print("Mundur: " + " <-> ".join(hasil))

    def insert_last(self, data):
        baru = DNode(data)
        if self.is_empty():
            self.head = self.tail = baru
        else:
            baru.prev = self.tail     # 1. baru menunjuk tail lama (mundur)
            self.tail.next = baru     # 2. tail lama menunjuk baru (maju)
            self.tail = baru          # 3. tail pindah ke baru

    def delete_front(self):
        if self.is_empty():
            print("List kosong")
            return
        if self.head is self.tail:    # hanya 1 node
            self.head = self.tail = None
        else:
            self.head = self.head.next
            self.head.prev = None     # putuskan pointer mundur ke node lama


dll = DoublyLinkedList()
for x in [10, 20, 30]:
    dll.insert_last(x)

dll.delete_front()
dll.display_forward()        # Output: Maju  : 20 <-> 30
dll.display_backward()       # Output: Mundur: 30 <-> 20
````

## ❌ 9. Menghapus Node Terakhir (`delete_last`)

### Konsep

| Baris | Penjelasan |

````python
class DNode:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.prev = None          # pointer ke node SEBELUMNYA
        self.next = None          # pointer ke node BERIKUTNYA


class DoublyLinkedList:
    def __init__(self):
        self.head = None              # node pertama
        self.tail = None              # node terakhir

    def is_empty(self):
        return self.head is None

    def display_forward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head               # mulai dari depan
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next            # maju lewat next
        print("Maju  : " + " <-> ".join(hasil))

    def display_backward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.tail               # mulai dari belakang
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.prev            # mundur lewat prev
        print("Mundur: " + " <-> ".join(hasil))

    def insert_last(self, data):
        baru = DNode(data)
        if self.is_empty():
            self.head = self.tail = baru
        else:
            baru.prev = self.tail     # 1. baru menunjuk tail lama (mundur)
            self.tail.next = baru     # 2. tail lama menunjuk baru (maju)
            self.tail = baru          # 3. tail pindah ke baru

    def delete_last(self):
        if self.is_empty():
            print("List kosong")
            return
        if self.head is self.tail:
            self.head = self.tail = None
        else:
            self.tail = self.tail.prev
            self.tail.next = None     # putuskan pointer maju ke node lama


dll = DoublyLinkedList()
for x in [10, 20, 30]:
    dll.insert_last(x)

dll.delete_last()
dll.display_forward()        # Output: Maju  : 10 <-> 20
dll.display_backward()       # Output: Mundur: 20 <-> 10
````

## 🎯 10. Menghapus Berdasarkan Nilai (`delete_value`)

### Konsep

Cari node yang bernilai `data`. Karena node punya `prev`, kita **tidak perlu berhenti di node sebelumnya**. Ada tiga kasus: node di `head`, node di `tail`, atau node di tengah (tetangga kiri dan kanan saling disambungkan).

```
delete_value(20):

[10]<->[20]<->[30]        [10]<--------->[30]
        cur               ([20] terlepas)
```

| Baris | Penjelasan |
|---|---|
| `cur is self.head` | Kasus depan: pakai `delete_front()` |
| `cur is self.tail` | Kasus belakang: pakai `delete_last()` |
| `cur.prev.next = cur.next` | Tetangga kiri melompati `cur` |
| `cur.next.prev = cur.prev` | Tetangga kanan melompati `cur` |

````python
class DNode:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.prev = None          # pointer ke node SEBELUMNYA
        self.next = None          # pointer ke node BERIKUTNYA


class DoublyLinkedList:
    def __init__(self):
        self.head = None              # node pertama
        self.tail = None              # node terakhir

    def is_empty(self):
        return self.head is None

    def display_forward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head               # mulai dari depan
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next            # maju lewat next
        print("Maju  : " + " <-> ".join(hasil))

    def display_backward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.tail               # mulai dari belakang
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.prev            # mundur lewat prev
        print("Mundur: " + " <-> ".join(hasil))

    def insert_last(self, data):
        baru = DNode(data)
        if self.is_empty():
            self.head = self.tail = baru
        else:
            baru.prev = self.tail     # 1. baru menunjuk tail lama (mundur)
            self.tail.next = baru     # 2. tail lama menunjuk baru (maju)
            self.tail = baru          # 3. tail pindah ke baru

    def delete_front(self):
        if self.is_empty():
            print("List kosong")
            return
        if self.head is self.tail:    # hanya 1 node
            self.head = self.tail = None
        else:
            self.head = self.head.next
            self.head.prev = None     # putuskan pointer mundur ke node lama

    def delete_last(self):
        if self.is_empty():
            print("List kosong")
            return
        if self.head is self.tail:
            self.head = self.tail = None
        else:
            self.tail = self.tail.prev
            self.tail.next = None     # putuskan pointer maju ke node lama

    def delete_value(self, data):
        cur = self.head
        while cur is not None and cur.data != data:
            cur = cur.next
        if cur is None:
            print(f"{data} tidak ditemukan")
            return
        if cur is self.head:
            self.delete_front()
        elif cur is self.tail:
            self.delete_last()
        else:
            cur.prev.next = cur.next  # tetangga kiri melompati cur
            cur.next.prev = cur.prev  # tetangga kanan melompati cur


dll = DoublyLinkedList()
for x in [10, 20, 30, 40]:
    dll.insert_last(x)

dll.delete_value(20)
dll.display_forward()        # Output: Maju  : 10 <-> 30 <-> 40
dll.display_backward()       # Output: Mundur: 40 <-> 30 <-> 10

dll.delete_value(10)         # kasus head
dll.delete_value(40)         # kasus tail
dll.display_forward()        # Output: Maju  : 30
dll.delete_value(99)         # Output: 99 tidak ditemukan
````

````python
class DNode:
    def __init__(self, data):
        self.data = data          # nilai yang disimpan
        self.prev = None          # pointer ke node SEBELUMNYA
        self.next = None          # pointer ke node BERIKUTNYA


class DoublyLinkedList:
    def __init__(self):
        self.head = None              # node pertama
        self.tail = None              # node terakhir

    def is_empty(self):
        return self.head is None

    def insert_front(self, data):
        baru = DNode(data)
        if self.is_empty():           # list kosong: head dan tail sama-sama node baru
            self.head = self.tail = baru
        else:
            baru.next = self.head     # 1. baru menunjuk head lama
            self.head.prev = baru     # 2. head lama menunjuk balik ke baru
            self.head = baru          # 3. head pindah ke baru

    def insert_last(self, data):
        baru = DNode(data)
        if self.is_empty():
            self.head = self.tail = baru
        else:
            baru.prev = self.tail     # 1. baru menunjuk tail lama (mundur)
            self.tail.next = baru     # 2. tail lama menunjuk baru (maju)
            self.tail = baru          # 3. tail pindah ke baru

    def insert_after(self, target, data):
        cur = self.head
        while cur is not None and cur.data != target:
            cur = cur.next
        if cur is None:
            print(f"{target} tidak ditemukan")
            return
        if cur is self.tail:          # target adalah node terakhir
            self.insert_last(data)
            return
        baru = DNode(data)
        baru.prev = cur               # 1. baru menunjuk mundur ke target
        baru.next = cur.next          # 2. baru menunjuk maju ke node setelah target
        cur.next.prev = baru          # 3. node setelah target menunjuk mundur ke baru
        cur.next = baru               # 4. target menunjuk maju ke baru

    def delete_front(self):
        if self.is_empty():
            print("List kosong")
            return
        if self.head is self.tail:    # hanya 1 node
            self.head = self.tail = None
        else:
            self.head = self.head.next
            self.head.prev = None     # putuskan pointer mundur ke node lama

    def delete_last(self):
        if self.is_empty():
            print("List kosong")
            return
        if self.head is self.tail:
            self.head = self.tail = None
        else:
            self.tail = self.tail.prev
            self.tail.next = None     # putuskan pointer maju ke node lama

    def delete_value(self, data):
        cur = self.head
        while cur is not None and cur.data != data:
            cur = cur.next
        if cur is None:
            print(f"{data} tidak ditemukan")
            return
        if cur is self.head:
            self.delete_front()
        elif cur is self.tail:
            self.delete_last()
        else:
            cur.prev.next = cur.next  # tetangga kiri melompati cur
            cur.next.prev = cur.prev  # tetangga kanan melompati cur

    def display_forward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head               # mulai dari depan
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next            # maju lewat next
        print("Maju  : " + " <-> ".join(hasil))

    def display_backward(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.tail               # mulai dari belakang
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.prev            # mundur lewat prev
        print("Mundur: " + " <-> ".join(hasil))


if __name__ == "__main__":
    dll = DoublyLinkedList()

    print("=== Insert ===")
    dll.insert_last(20)
    dll.insert_last(30)
    dll.insert_front(10)
    dll.insert_after(20, 25)
    dll.display_forward()        # Output: Maju  : 10 <-> 20 <-> 25 <-> 30
    dll.display_backward()       # Output: Mundur: 30 <-> 25 <-> 20 <-> 10

    print("\n=== Delete ===")
    dll.delete_front()
    dll.delete_last()
    dll.delete_value(25)
    dll.display_forward()        # Output: Maju  : 20
````

## 🔑 Operasi Penting yang Wajib Dipahami

| Operasi | Cara Kerja | Kompleksitas |
|---|---|---|
| `insert_front(x)` | Atur `next` baru dan `prev` head lama | O(1) |
| `insert_last(x)` | Pakai `tail`, tanpa penelusuran | O(1) |
| `insert_after(t, x)` | Cari target, atur 4 pointer | O(n) |
| `delete_front()` | `head = head.next`, `head.prev = None` | O(1) |
| `delete_last()` | `tail = tail.prev`, `tail.next = None` | O(1) |
| `delete_value(x)` | Cari node, sambungkan tetangganya | O(n) |
| `display_forward/backward()` | Telusuri lewat `next` / `prev` | O(n) |

### Singly vs Doubly Linked List

| | Singly | Doubly |
|---|---|---|
| Pointer per node | `next` | `prev` dan `next` |
| Arah penelusuran | Maju saja | Maju dan mundur |
| `insert_last` | O(n) | O(1) dengan `tail` |
| `delete_last` | O(n) | O(1) dengan `tail` |
| Menghapus node tertentu | Butuh node sebelumnya | Cukup node itu sendiri |
| Memori | Lebih hemat | Lebih besar (satu pointer ekstra) |

> 🧠 **Tips kunci:** Setiap kali menyisipkan atau menghapus, pastikan **kedua arah** pointer konsisten (`next` dan `prev`). Selalu tangani kasus: list kosong, satu node, `head`, dan `tail`. Jangan lupa memperbarui `tail` (dan `head`) saat node di ujung berubah.
