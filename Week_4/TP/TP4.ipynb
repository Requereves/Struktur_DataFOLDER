## Apa Itu Linked List?

**Linked List** adalah struktur data berupa kumpulan **node** yang saling terhubung satu sama lain melalui **pointer** (penunjuk).

Berbeda dengan array/list biasa di Python yang menyimpan data secara berurutan di memori, Linked List menyimpan setiap elemen di tempat yang berbeda-beda di memori, lalu menghubungkannya menggunakan pointer.

```
[Node 1] --next--> [Node 2] --next--> [Node 3] --next--> None
  data=10            data=20            data=30
```

Setiap **node** terdiri dari dua bagian:
- `data` → nilai yang disimpan
- `next` → pointer yang menunjuk ke node berikutnya (atau `None` jika node terakhir)

**`head`** adalah pointer khusus yang selalu menunjuk ke node **pertama** dalam linked list. Jika `head = None`, berarti linked list **kosong**.

```
--head--> [Node 1]
--head--> None
```

## 📦 1. Membuat Node dan Menginisialisasi `head`

### Konsep
Sebelum bisa melakukan apa pun, kita perlu membuat "cetakan" node menggunakan `class`, lalu membuat `head` sebagai pintu masuk ke linked list kita.

````python
# Definisi kelas Node
class Node:
    def __init__(self, data):   # Konstruktor: dipanggil saat Node baru dibuat
        self.data = data        # Menyimpan nilai/data dari node ini
        self.next = None        # Pointer ke node berikutnya, awalnya None (belum ada)

# Definisi kelas LinkedList
class LinkedList:
    def __init__(self):         # Konstruktor LinkedList
        self.head = None        # head awalnya None, artinya linked list masih kosong
````

| Baris | Penjelasan |
|---|---|
| `class Node:` | Membuat cetak biru (blueprint) untuk setiap node |
| `def __init__(self, data):` | Fungsi yang otomatis dipanggil saat `Node(nilai)` dibuat |
| `self.data = data` | Menyimpan nilai yang diberikan ke dalam node |
| `self.next = None` | Node baru belum terhubung ke mana-mana |
| `class LinkedList:` | Blueprint untuk linked list itu sendiri |
| `self.head = None` | Awalnya linked list kosong, head tidak menunjuk ke mana-mana |

### Cara Pakai & Output

````python
ll = LinkedList()
print(ll.head)   # Output: None
````

## ✅ 2. Memeriksa Apakah Linked List Kosong (`isEmpty`)

### Konsep
Sebelum melakukan operasi seperti delete atau traversal, kita perlu tahu apakah linked list punya isi atau tidak. Caranya sederhana: cek apakah `head` bernilai `None`.

````python
class Node:
    def __init__(self, data):   
        self.data = data        
        self.next = None        

class LinkedList:
    def __init__(self):         
        self.head = None        
    
    def is_empty(self):
        return self.head is None   # Mengembalikan True jika head None (kosong), False jika tidak
````

````python
ll = LinkedList()
print(ll.is_empty())   # Output: True

# Setelah ada isi (contoh nanti), hasilnya akan False
````

## ➕ 3. Insert di Awal (`insert_at_beginning`)

### Konsep
Menambahkan node **baru** di posisi **paling depan** linked list. Node baru langsung menjadi `head`.

Langkah-langkahnya:
1. Buat node baru
2. Arahkan `next` node baru ke `head` lama
3. Jadikan node baru sebagai `head` yang baru

```
Sebelum: head --> [20] --> [30] --> None
Sesudah: head --> [10] --> [20] --> [30] --> None
                  ^baru
```

````python
class Node:
    def __init__(self, data):   
        self.data = data        
        self.next = None        

class LinkedList:
    def __init__(self):         
        self.head = None        
    
    def insert_at_beginning(self, data):
        new_node = Node(data)        # Buat node baru dengan data yang diberikan
        new_node.next = self.head    # Node baru menunjuk ke head lama (node pertama sebelumnya)
        self.head = new_node         # head sekarang menunjuk ke node baru (jadikan node pertama)
````

| Baris | Penjelasan |
|---|---|
| `new_node = Node(data)` | Membuat node baru di memori |
| `new_node.next = self.head` | Sambungkan node baru ke node yang sebelumnya pertama |
| `self.head = new_node` | Pindahkan head agar menunjuk ke node baru |

````python
ll = LinkedList()
ll.insert_at_beginning(30)
ll.insert_at_beginning(20)
ll.insert_at_beginning(10)
# Linked list: 10 -> 20 -> 30 -> None
````

## ➕ 4. Insert di Akhir (`insert_at_end`)

### Konsep
Menambahkan node baru di posisi **paling belakang**. Kita harus **berjalan** dari `head` sampai node terakhir (yang `next`-nya `None`), lalu sambungkan node baru di sana.

```
Sebelum: head --> [10] --> [20] --> None
Sesudah: head --> [10] --> [20] --> [30] --> None
                                    ^baru
```

````python
class Node:
    def __init__(self, data):   
        self.data = data        
        self.next = None        

class LinkedList:
    def __init__(self):         
        self.head = None        

    def insert_at_end(self, data):
        new_node = Node(data)        # Buat node baru

        if self.head is None:        # Jika linked list masih kosong...
            self.head = new_node     # ...langsung jadikan node baru sebagai head
            return                   # Selesai, tidak perlu lanjut

        current = self.head          # Mulai dari node pertama (head)
        while current.next is not None:   # Selama masih ada node berikutnya...
            current = current.next        # ...pindah ke node berikutnya
        # Saat loop selesai, current berada di node TERAKHIR
        current.next = new_node      # Sambungkan node terakhir ke node baru
````

| Baris | Penjelasan |
|---|---|
| `if self.head is None:` | Kasus khusus: kalau list kosong, node baru langsung jadi head |
| `current = self.head` | Mulai penelusuran dari awal |
| `while current.next is not None:` | Terus berjalan selama bukan node terakhir |
| `current = current.next` | Maju ke node berikutnya |
| `current.next = new_node` | Sambungkan node terakhir ke node baru |

````python
ll = LinkedList()
ll.insert_at_end(10)
ll.insert_at_end(20)
ll.insert_at_end(30)
# Linked list: 10 -> 20 -> 30 -> None
````

## ➕ 5. Insert di Posisi Tertentu (`insert_at_position`)

### Konsep
Menyisipkan node baru di **posisi ke-n** (mulai dari 0). Kita perlu berjalan sampai posisi sebelumnya, lalu menyambungkan node baru di antara dua node.

```
Sisipkan 99 di posisi 1:
Sebelum: head --> [10] --> [20] --> [30] --> None
Sesudah: head --> [10] --> [99] --> [20] --> [30] --> None
```

````python
class Node:
    def __init__(self, data):   
        self.data = data        
        self.next = None        

class LinkedList:
    def __init__(self):         
        self.head = None        

    def insert_at_end(self, data):
        new_node = Node(data)        

        if self.head is None:        
            self.head = new_node     
            return                   

        current = self.head          
        while current.next is not None:
            current = current.next     
        current.next = new_node      

    def insert_at_position(self, data, position):
        new_node = Node(data)          # Buat node baru

        if position == 0:              # Jika posisi 0, sama seperti insert di awal
            new_node.next = self.head
            self.head = new_node
            return

        current = self.head            # Mulai dari head
        current_position = 0           # Lacak posisi saat ini

        while current is not None and current_position < position - 1:
            # Berjalan sampai satu posisi SEBELUM posisi tujuan
            current = current.next
            current_position += 1

        if current is None:            # Jika posisi melebihi panjang list
            print("Posisi melebihi panjang linked list!")
            return

        new_node.next = current.next   # Node baru menunjuk ke node yang sebelumnya ada di posisi tujuan
        current.next = new_node        # Node sebelum posisi tujuan menunjuk ke node baru
````

### Penjelasan Penting: Urutan Pointer

> ⚠️ **Urutan ini SANGAT penting!** Harus `new_node.next = current.next` dulu, BARU `current.next = new_node`. Kalau dibalik, kita akan kehilangan node-node setelahnya karena pointer sudah ditimpa.

````python
ll = LinkedList()
ll.insert_at_end(10)
ll.insert_at_end(20)
ll.insert_at_end(30)
ll.insert_at_position(99, 1)
# Linked list: 10 -> 99 -> 20 -> 30 -> None
````

## 🔍 6. Traversal / Display Seluruh Node

### Konsep
**Traversal** artinya mengunjungi setiap node satu per satu dari `head` hingga node terakhir (`None`), biasanya untuk menampilkan isinya.

````python
class Node:
    def __init__(self, data):   
        self.data = data        
        self.next = None        

class LinkedList:
    def __init__(self):         
        self.head = None          

    def insert_at_end(self, data):
        new_node = Node(data)        

        if self.head is None:        
            self.head = new_node     
            return                   

        current = self.head          
        while current.next is not None:
            current = current.next     
        current.next = new_node      

    def display(self):
        if self.head is None:          # Cek jika list kosong
            print("Linked list kosong")
            return

        current = self.head            # Mulai dari head
        result = []                    # Tampung data setiap node

        while current is not None:     # Selama masih ada node...
            result.append(str(current.data))   # Tambahkan data node ke list result
            current = current.next     # Pindah ke node berikutnya

        print(" -> ".join(result) + " -> None")  # Tampilkan semua node
````

````python
ll = LinkedList()
ll.insert_at_end(10)
ll.insert_at_end(20)
ll.insert_at_end(30)
ll.display()
````

## 🔍 7. Searching Berdasarkan Nilai

### Konsep
Mencari apakah suatu nilai ada di dalam linked list. Kita berjalan dari `head` sambil membandingkan setiap `data` node dengan nilai yang dicari.

````python
class Node:
    def __init__(self, data):   
        self.data = data        
        self.next = None        

class LinkedList:
    def __init__(self):         
        self.head = None        

    def insert_at_end(self, data):
        new_node = Node(data)        

        if self.head is None:        
            self.head = new_node     
            return                   

        current = self.head          
        while current.next is not None:
            current = current.next     
        current.next = new_node      

    def search(self, target):
        current = self.head            # Mulai dari head
        position = 0                   # Lacak posisi

        while current is not None:     # Selama masih ada node
            if current.data == target: # Jika data node sama dengan yang dicari
                print(f"Data {target} ditemukan di posisi {position}")
                return True            # Kembalikan True (ditemukan)
            current = current.next     # Pindah ke node berikutnya
            position += 1

        print(f"Data {target} tidak ditemukan")
        return False                   # Kembalikan False (tidak ditemukan)
````

````python
ll = LinkedList()
ll.insert_at_end(10)
ll.insert_at_end(20)
ll.insert_at_end(30)

ll.search(20)   # Ada
ll.search(99)   # Tidak ada
````

## 🔢 8. Menghitung Jumlah Node (`count`)

### Konsep
Menghitung berapa banyak node yang ada. Caranya: traversal dari awal sampai akhir sambil menghitung.

````python
class Node:
    def __init__(self, data):   
        self.data = data        
        self.next = None        

class LinkedList:
    def __init__(self):         
        self.head = None        

    def insert_at_end(self, data):
        new_node = Node(data)        

        if self.head is None:        
            self.head = new_node     
            return                   

        current = self.head          
        while current.next is not None:
            current = current.next     
        current.next = new_node      
             
    def count(self):
        counter = 0                    # Mulai dari 0
        current = self.head            # Mulai dari head

        while current is not None:     # Selama masih ada node...
            counter += 1               # Tambah counter
            current = current.next     # Pindah ke node berikutnya

        return counter                 # Kembalikan jumlah node    
````

````python
ll = LinkedList()
ll.insert_at_end(10)
ll.insert_at_end(20)
ll.insert_at_end(30)
print("Jumlah node:", ll.count())
````

## 🔄 9. Update Nilai Node

### Konsep
Mengubah nilai (`data`) dari node di posisi tertentu. Lakukan traversal ke posisi yang dituju, lalu ubah nilainya.

````python
class Node:
    def __init__(self, data):   
        self.data = data        
        self.next = None        

class LinkedList:
    def __init__(self):         
        self.head = None          

    def insert_at_end(self, data):
        new_node = Node(data)        

        if self.head is None:        
            self.head = new_node     
            return                   

        current = self.head          
        while current.next is not None:
            current = current.next     
        current.next = new_node         

    def display(self):
        if self.head is None:         
            print("Linked list kosong")
            return

        current = self.head            
        result = []                    

        while current is not None:     
            result.append(str(current.data))   
            current = current.next     

        print(" -> ".join(result) + " -> None") 
    
    def update(self, position, new_data):
        current = self.head            # Mulai dari head
        current_position = 0           # Posisi saat ini

        while current is not None:     # Selama masih ada node
            if current_position == position:    # Jika posisi sesuai target
                current.data = new_data         # Ubah data node tersebut
                print(f"Node di posisi {position} diubah menjadi {new_data}")
                return
            current = current.next     # Pindah ke node berikutnya
            current_position += 1

        print("Posisi tidak ditemukan!")   # Jika posisi melebihi panjang list
````

````python
ll = LinkedList()
ll.insert_at_end(10)
ll.insert_at_end(20)
ll.insert_at_end(30)
ll.display()        # Sebelum update
ll.update(1, 99)
ll.display()        # Sesudah update
````

## ❌ 10. Delete di Awal (`delete_at_beginning`)

### Konsep
Menghapus node **pertama**. Cukup geser `head` ke node kedua.

```
Sebelum: head --> [10] --> [20] --> [30] --> None
Sesudah: head --> [20] --> [30] --> None
```

````python
class Node:
    def __init__(self, data):   
        self.data = data        
        self.next = None        

class LinkedList:
    def __init__(self):         
        self.head = None           

    def insert_at_end(self, data):
        new_node = Node(data)        

        if self.head is None:        
            self.head = new_node     
            return                   

        current = self.head          
        while current.next is not None:
            current = current.next     
        current.next = new_node             

    def display(self):
        if self.head is None:         
            print("Linked list kosong")
            return

        current = self.head            
        result = []                    

        while current is not None:     
            result.append(str(current.data))   
            current = current.next     

        print(" -> ".join(result) + " -> None")  

    def delete_at_beginning(self):
        if self.head is None:          # Kasus khusus: list kosong
            print("Linked list kosong, tidak ada yang dihapus")
            return

        deleted_data = self.head.data  # Simpan data yang akan dihapus (untuk info)
        self.head = self.head.next     # Geser head ke node kedua (node pertama "terlepas")
        print(f"Data {deleted_data} berhasil dihapus dari awal")
````

````python
ll = LinkedList()
ll.insert_at_end(10)
ll.insert_at_end(20)
ll.insert_at_end(30)
ll.delete_at_beginning()
ll.display()
````

## ❌ 11. Delete di Akhir (`delete_at_end`)

### Konsep
Menghapus node **terakhir**. Kita harus berjalan sampai node **kedua dari belakang**, lalu set `next`-nya ke `None`.

```
Sebelum: head --> [10] --> [20] --> [30] --> None
Sesudah: head --> [10] --> [20] --> None
```

````python
class Node:
    def __init__(self, data):   
        self.data = data        
        self.next = None        

class LinkedList:
    def __init__(self):         
        self.head = None     

    def insert_at_end(self, data):
        new_node = Node(data)        

        if self.head is None:        
            self.head = new_node     
            return                   

        current = self.head          
        while current.next is not None:
            current = current.next     
        current.next = new_node                              

    def display(self):
        if self.head is None:         
            print("Linked list kosong")
            return

        current = self.head            
        result = []                    

        while current is not None:     
            result.append(str(current.data))   
            current = current.next     

        print(" -> ".join(result) + " -> None")                    

    def delete_at_end(self):
        if self.head is None:               # Kasus: list kosong
            print("Linked list kosong!")
            return

        if self.head.next is None:          # Kasus: hanya ada 1 node
            deleted_data = self.head.data
            self.head = None                # Hapus satu-satunya node, list jadi kosong
            print(f"Data {deleted_data} berhasil dihapus")
            return

        current = self.head                 # Mulai dari head
        while current.next.next is not None:   # Berhenti saat current adalah node kedua dari belakang
            current = current.next

        deleted_data = current.next.data    # Simpan data node terakhir
        current.next = None                 # Putus sambungan ke node terakhir
        print(f"Data {deleted_data} berhasil dihapus dari akhir")
````

### Penjelasan Penting

> `current.next.next is not None` — artinya: "masih ada node dua langkah ke depan". Ketika kondisi ini False, berarti `current.next` adalah node terakhir, dan `current` adalah node kedua dari belakang.

````python
ll = LinkedList()
ll.insert_at_end(10)
ll.insert_at_end(20)
ll.insert_at_end(30)
ll.delete_at_end()
ll.display()
````

## ❌ 12. Delete Berdasarkan Nilai (`delete_by_value`)

### Konsep
Menghapus node yang memiliki **nilai tertentu**. Kita perlu menyimpan referensi ke node **sebelumnya** (`prev`) agar bisa menyambungkan kembali.

```
Hapus node dengan nilai 20:
Sebelum: head --> [10] --> [20] --> [30] --> None
                   prev    current
Sesudah: head --> [10] --> [30] --> None
                   prev.next = current.next
```

````python
class Node:
    def __init__(self, data):   
        self.data = data        
        self.next = None        

class LinkedList:
    def __init__(self):         
        self.head = None     

    def insert_at_end(self, data):
        new_node = Node(data)        

        if self.head is None:        
            self.head = new_node     
            return                   

        current = self.head          
        while current.next is not None:
            current = current.next     
        current.next = new_node                              

    def display(self):
        if self.head is None:         
            print("Linked list kosong")
            return

        current = self.head            
        result = []                    

        while current is not None:     
            result.append(str(current.data))   
            current = current.next     

        print(" -> ".join(result) + " -> None")                    

    def delete_by_value(self, target):
        if self.head is None:               # Kasus: list kosong
            print("Linked list kosong!")
            return

        if self.head.data == target:        # Kasus khusus: node yang dihapus adalah head
            self.head = self.head.next      # Geser head ke node berikutnya
            print(f"Data {target} berhasil dihapus")
            return

        current = self.head                 # Mulai dari head
        prev = None                         # Simpan node sebelumnya (awalnya tidak ada)

        while current is not None:          # Selama masih ada node
            if current.data == target:      # Jika data ditemukan
                prev.next = current.next    # Sambungkan node sebelumnya langsung ke node sesudahnya
                print(f"Data {target} berhasil dihapus")
                return
            prev = current                  # Simpan node saat ini sebagai 'sebelumnya'
            current = current.next          # Maju ke node berikutnya

        print(f"Data {target} tidak ditemukan!")  # Jika tidak ketemu
````

````python
ll = LinkedList()
ll.insert_at_end(10)
ll.insert_at_end(20)
ll.insert_at_end(30)
ll.delete_by_value(20)
ll.display()
ll.delete_by_value(99)   # Nilai yang tidak ada
````

## 🧩 Kode Lengkap (Semua Operasi Dalam Satu Class)

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    # Cek apakah kosong
    def is_empty(self):
        return self.head is None

    # Tampilkan semua node
    def display(self):
        if self.head is None:
            print("Linked list kosong")
            return
        current = self.head
        result = []
        while current is not None:
            result.append(str(current.data))
            current = current.next
        print(" -> ".join(result) + " -> None")

    # Hitung jumlah node
    def count(self):
        counter = 0
        current = self.head
        while current is not None:
            counter += 1
            current = current.next
        return counter

    # Insert di awal
    def insert_at_beginning(self, data):
        new_node = Node(data)
        new_node.next = self.head
        self.head = new_node

    # Insert di akhir
    def insert_at_end(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        current = self.head
        while current.next is not None:
            current = current.next
        current.next = new_node

    # Insert di posisi tertentu
    def insert_at_position(self, data, position):
        new_node = Node(data)
        if position == 0:
            new_node.next = self.head
            self.head = new_node
            return
        current = self.head
        current_position = 0
        while current is not None and current_position < position - 1:
            current = current.next
            current_position += 1
        if current is None:
            print("Posisi melebihi panjang linked list!")
            return
        new_node.next = current.next
        current.next = new_node

    # Delete di awal
    def delete_at_beginning(self):
        if self.head is None:
            print("Linked list kosong!")
            return
        deleted_data = self.head.data
        self.head = self.head.next
        print(f"Data {deleted_data} berhasil dihapus dari awal")

    # Delete di akhir
    def delete_at_end(self):
        if self.head is None:
            print("Linked list kosong!")
            return
        if self.head.next is None:
            deleted_data = self.head.data
            self.head = None
            print(f"Data {deleted_data} berhasil dihapus")
            return
        current = self.head
        while current.next.next is not None:
            current = current.next
        deleted_data = current.next.data
        current.next = None
        print(f"Data {deleted_data} berhasil dihapus dari akhir")

    # Delete berdasarkan nilai
    def delete_by_value(self, target):
        if self.head is None:
            print("Linked list kosong!")
            return
        if self.head.data == target:
            self.head = self.head.next
            print(f"Data {target} berhasil dihapus")
            return
        current = self.head
        prev = None
        while current is not None:
            if current.data == target:
                prev.next = current.next
                print(f"Data {target} berhasil dihapus")
                return
            prev = current
            current = current.next
        print(f"Data {target} tidak ditemukan!")

    # Searching berdasarkan nilai
    def search(self, target):
        current = self.head
        position = 0
        while current is not None:
            if current.data == target:
                print(f"Data {target} ditemukan di posisi {position}")
                return True
            current = current.next
            position += 1
        print(f"Data {target} tidak ditemukan")
        return False

    # Update nilai node
    def update(self, position, new_data):
        current = self.head
        current_position = 0
        while current is not None:
            if current_position == position:
                current.data = new_data
                print(f"Node di posisi {position} diubah menjadi {new_data}")
                return
            current = current.next
            current_position += 1
        print("Posisi tidak ditemukan!")
````

````python
# ===================== DEMO =====================
if __name__ == "__main__":
    ll = LinkedList()

    print("=== Insert ===")
    ll.insert_at_end(10)
    ll.insert_at_end(20)
    ll.insert_at_end(30)
    ll.insert_at_beginning(5)
    ll.insert_at_position(15, 2)
    ll.display()                        # 5 -> 10 -> 15 -> 20 -> 30 -> None
    print("Jumlah node:", ll.count())   # 5

    print("\n=== Search ===")
    ll.search(15)
    ll.search(99)

    print("\n=== Update ===")
    ll.update(2, 99)
    ll.display()

    print("\n=== Delete ===")
    ll.delete_at_beginning()
    ll.display()
    ll.delete_at_end()
    ll.display()
    ll.delete_by_value(20)
    ll.display()
````

## 📐 Konsep Pointer: Ringkasan Visual

```
KONDISI AWAL (list kosong):
head --> None

SETELAH insert_at_end(10):
head --> [10|None]

SETELAH insert_at_end(20):
head --> [10|next] --> [20|None]

SETELAH insert_at_beginning(5):
head --> [5|next] --> [10|next] --> [20|None]

SAAT DELETE node [10]:
  prev = [5], current = [10], current.next = [20]
  prev.next = current.next   -->   [5|next] --> [20|None]
```

## 🔑 Pointer Penting yang Wajib Dipahami

| Situasi | Yang dilakukan pointer |
|---|---|
| Insert di awal | `new.next = head` → `head = new` |
| Insert di akhir | Cari node `next=None` → `node.next = new` |
| Insert di tengah | `new.next = current.next` → `current.next = new` |
| Delete di awal | `head = head.next` |
| Delete di akhir | Cari node kedua dari akhir → `node.next = None` |
| Delete di tengah | `prev.next = current.next` |

> 🧠 **Tips kunci:** Selalu pikirkan "node mana yang perlu ditunjuk ulang?" sebelum menulis kode. Gambar dulu di kertas jika perlu!

## 🏋️ Soal Latihan

### Soal 1 — Membalik Linked List (`reverse`)

**Deskripsi:**  
Buatlah method `reverse(self)` yang membalik urutan seluruh node dalam linked list **tanpa membuat node baru**. Hanya boleh mengubah arah pointer.

**Contoh:**
```
Input : 1 -> 2 -> 3 -> 4 -> 5 -> None
Output: 5 -> 4 -> 3 -> 2 -> 1 -> None
```

**Petunjuk:**  
Gunakan tiga variabel bantu: `prev`, `current`, dan `next_node`. Di setiap iterasi, balik arah pointer `current.next` agar menunjuk ke `prev`.

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class LinkedList:
    def __init__(self):
        self.head = None

    def insert_at_end(self, data):
        new_node = Node(data)
        if self.head is None:
            self.head = new_node
            return
        current = self.head
        while current.next is not None:
            current = current.next
        current.next = new_node

    def display(self):
        if self.head is None:
            print("Linked list kosong")
            return
        current = self.head
        result = []
        while current is not None:
            result.append(str(current.data))
            current = current.next
        print(" -> ".join(result) + " -> None")

    def reverse(self):
        prev = None             # Awalnya tidak ada node sebelumnya
        current = self.head     # Mulai dari head

        while current is not None:
            next_node = current.next    # TODO 1: Simpan node berikutnya sebelum pointer diubah

            # TODO 2: Balik arah pointer current agar menunjuk ke prev
            current.next = prev

            # TODO 3: Geser prev ke posisi current
            prev = current

            # TODO 4: Geser current ke node berikutnya (yang sudah disimpan)
            current = next_node

        # TODO 5: Setelah loop selesai, prev berada di node terakhir (yang sekarang jadi kepala)
        # Ubah head agar menunjuk ke prev
        self.head = prev
````

````python
# ===== MAIN =====
ll = LinkedList()
for i in [1, 2, 3, 4, 5]:
    ll.insert_at_end(i)

print("Sebelum reverse:")
ll.display()

ll.reverse()

print("Sesudah reverse:")
ll.display()

# Output yang diharapkan:
# Sebelum reverse:
# 1 -> 2 -> 3 -> 4 -> 5 -> None
# Sesudah reverse:
# 5 -> 4 -> 3 -> 2 -> 1 -> None
````
