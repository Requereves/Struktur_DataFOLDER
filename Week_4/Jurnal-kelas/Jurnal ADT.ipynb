## Bagian 1: Apa itu Class dan Object?

Sebelum masuk ke ADT, kamu harus paham dulu apa itu **class** dan **object**.

**Class** adalah cetakan atau blueprint. Seperti desain rumah di atas kertas.

**Object** adalah hasil dari cetakan itu. Seperti rumah yang sudah jadi dibangun dari desain tersebut.

````python
# Class = cetakan
class Hewan:
    def __init__(self, nama):
        self.nama = nama

# Object = hasil cetakan
kucing = Hewan("Kucing")   # objek pertama
anjing = Hewan("Anjing")   # objek kedua, berbeda dari yang pertama
````

`kucing` dan `anjing` adalah dua objek yang berbeda, tapi sama-sama dibuat dari class `Hewan`.

## Bagian 2: Access Modifier — Public, Protected, Private

Ini adalah aturan **siapa yang boleh mengakses** suatu atribut atau method.

Bayangkan kamu punya brankas di rumah:
- **Laci meja** (public) — siapa saja boleh buka
- **Lemari kamar** (protected) — idealnya hanya keluarga
- **Brankas** (private) — hanya kamu sendiri yang boleh buka

### 2.1 Public

Tidak ada tanda khusus. Bisa diakses dari mana saja — dari dalam class, dari luar class, dari class lain.

````python
class Mahasiswa:
    def __init__(self, nama):
        self.nama = nama        # PUBLIC — tanpa underscore

mhs = Mahasiswa("Budi")
print(mhs.nama)                 # Bisa diakses langsung dari luar
# Output: Budi

mhs.nama = "Andi"               # Bisa diubah langsung dari luar
print(mhs.nama)
# Output: Andi
````

### 2.2 Protected

Diawali **satu underscore** `_`. Ini adalah **konvensi** (kesepakatan) — Python tidak benar-benar memblokir aksesnya, tapi artinya "tolong jangan diakses langsung dari luar, ini untuk internal class".

````python
class Mahasiswa:
    def __init__(self, nama, ipk):
        self.nama = nama
        self._ipk = ipk         # PROTECTED — satu underscore

mhs = Mahasiswa("Budi", 3.5)
print(mhs._ipk)                 # Masih bisa diakses, tapi tidak disarankan
# Output: 3.5
````

### 2.3 Private

Diawali **dua underscore** `__`. Python benar-benar menyembunyikan atribut ini. Tidak bisa diakses langsung dari luar class.

````python
class Mahasiswa:
    def __init__(self, nama, password):
        self.nama = nama
        self.__password = password   # PRIVATE — dua underscore

mhs = Mahasiswa("Budi", "rahasia123")

print(mhs.nama)           # OK
# Output: Budi

print(mhs.__password)     # ERROR!
# AttributeError: 'Mahasiswa' object has no attribute '__password'
````

### Tabel Perbandingan

| Jenis | Penulisan | Diakses dari Luar? | Kapan Dipakai? |
|---|---|---|---|
| Public | `nama` | ✅ Bebas | Data yang memang boleh diakses siapa saja |
| Protected | `_nama` | ⚠️ Bisa tapi jangan | Data internal, hanya untuk class dan turunannya |
| Private | `__nama` | ❌ Tidak bisa | Data sensitif yang harus dilindungi |

### Lalu bagaimana cara mengakses atribut private?

Jawabannya: melalui **method** yang disediakan oleh class itu sendiri. Inilah inti dari ADT — data disembunyikan, tapi tetap bisa digunakan lewat pintu yang sudah disediakan.

````python
class Mahasiswa:
    def __init__(self, nama, password):
        self.nama = nama
        self.__password = password

    # Pintu resmi untuk membaca password
    def get_password(self):
        return self.__password

    # Pintu resmi untuk mengubah password
    def set_password(self, password_baru):
        self.__password = password_baru


mhs = Mahasiswa("Budi", "rahasia123")

# Akses lewat method — BENAR
print(mhs.get_password())       # Output: rahasia123

# Ubah lewat method — BENAR
mhs.set_password("passwordbaru")
print(mhs.get_password())       # Output: passwordbaru
````

## Bagian 3: Constructor — `__init__`

### Apa itu Constructor?

**Constructor** adalah method spesial yang **otomatis dipanggil** saat kamu membuat objek baru. Tugasnya adalah **menyiapkan** objek — mengisi semua atribut dengan nilai awal.

Di Python, constructor selalu bernama `__init__`.

````python
class Lampu:
    def __init__(self, warna, watt):   # Constructor
        self.__warna = warna
        self.__watt = watt
        self.__menyala = False          # nilai default

# Saat baris ini dijalankan, __init__ langsung dipanggil otomatis
lampu1 = Lampu("putih", 10)
````

### Parameter `self`

`self` adalah kata kunci yang merujuk ke **objek itu sendiri**. Setiap method dalam class wajib punya `self` sebagai parameter pertama.

Bayangkan `self` seperti kata "saya" — ketika kamu menulis `self.__warna`, artinya "warna milik saya (objek ini)".

````python
class Lampu:
    def __init__(self, warna, watt):
        # self.__warna = warna
        # artinya: simpan nilai 'warna' sebagai atribut 'warna' MILIK objek ini
        self.__warna = warna
        self.__watt = watt
````

### Default Parameter

Kamu bisa memberi nilai default pada parameter constructor. Jika tidak diisi saat membuat objek, nilai default yang dipakai.

````python
class Lampu:
    def __init__(self, warna, watt, menyala=False):
        #                               ^^^^^^^^^^
        #                          nilai default = False
        self.__warna = warna
        self.__watt = watt
        self.__menyala = menyala

lampu1 = Lampu("putih", 10)          # menyala = False (pakai default)
lampu2 = Lampu("kuning", 5, True)    # menyala = True (diisi manual)
````

## Bagian 4: Accessor — Method untuk Membaca Data

### Apa itu Accessor?

**Accessor** adalah method yang tugasnya **hanya membaca** data dari dalam object, lalu mengembalikannya. Accessor **tidak boleh mengubah** apapun.

Ciri-ciri accessor:
- Biasanya namanya diawali `get_` atau `is_`
- Selalu punya `return`
- Tidak mengubah atribut apapun

````python
class Lampu:
    def __init__(self, warna, watt):
        self.__warna = warna
        self.__watt = watt
        self.__menyala = False

    # ACCESSOR — hanya membaca, lalu return
    def get_warna(self) -> str:
        return self.__warna

    def get_watt(self) -> int:
        return self.__watt

    def is_menyala(self) -> bool:       # is_ untuk yang return True/False
        return self.__menyala


lampu1 = Lampu("putih", 10)
print(lampu1.get_warna())       # Output: putih
print(lampu1.get_watt())        # Output: 10
print(lampu1.is_menyala())      # Output: False
````

### Accessor yang Mengembalikan Dictionary

Kadang semua data dikembalikan sekaligus dalam bentuk dictionary.

````python
class Lampu:
    def __init__(self, warna, watt):
        self.__warna = warna
        self.__watt = watt
        self.__menyala = False

    # ACCESSOR — hanya membaca, lalu return
    def get_warna(self) -> str:
        return self.__warna

    def get_watt(self) -> int:
        return self.__watt

    def is_menyala(self) -> bool:       # is_ untuk yang return True/False
        return self.__menyala
    
    def get_data(self) -> dict:
        return {
            "warna"   : self.__warna,
            "watt"    : self.__watt,
            "menyala" : self.__menyala
        }

lampu1 = Lampu("putih", 10)
print(lampu1.get_warna())       # Output: putih
print(lampu1.get_watt())        # Output: 10
print(lampu1.is_menyala())      # Output: False

# Penggunaan:
print(lampu1.get_data())
# Output: {'warna': 'putih', 'watt': 10, 'menyala': False}
````

## Bagian 5: Mutator — Method untuk Mengubah Data

### Apa itu Mutator?

**Mutator** adalah method yang tugasnya **mengubah** nilai atribut di dalam object. Di sinilah biasanya ada **validasi** — pengecekan apakah perubahan boleh dilakukan atau tidak.

Ciri-ciri mutator:
- Biasanya namanya diawali `set_` atau kata kerja (`nyalakan`, `matikan`, `setor`, `tarik`)
- Mengubah nilai atribut
- Sering return `bool` (True jika berhasil, False jika gagal)

````python
class Lampu:
    def __init__(self, warna, watt):
        self.__warna = warna
        self.__watt = watt
        self.__menyala = False

    # MUTATOR sederhana — langsung ubah tanpa validasi
    def set_warna(self, warna_baru: str):
        self.__warna = warna_baru

    # MUTATOR dengan validasi — ubah hanya jika kondisi terpenuhi
    def nyalakan(self) -> bool:
        if not self.__menyala:
            self.__menyala = True
            print("Lampu dinyalakan.")
            return True
        else:
            print("Lampu sudah menyala.")
            return False

    def matikan(self) -> bool:
        if self.__menyala:
            self.__menyala = False
            print("Lampu dimatikan.")
            return True
        else:
            print("Lampu sudah mati.")
            return False

    def is_menyala(self) -> bool:
        return self.__menyala


lampu1 = Lampu("putih", 10)
lampu1.nyalakan()           # Output: Lampu dinyalakan.
lampu1.nyalakan()           # Output: Lampu sudah menyala.
lampu1.matikan()            # Output: Lampu dimatikan.
print(lampu1.is_menyala())  # Output: False
````
