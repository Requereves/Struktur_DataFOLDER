# Pekan 3 - Singly Linked List (Operasi Lanjutan)

## Operasi Lanjutan pada Singly Linked List

Di Pekan 2 kita sudah membuat `insert_front`, `insert_last`, `display`, `length`, dan `search`. Di pekan ini kita menambah operasi yang lebih sering muncul di ujian: **menyisipkan di tengah**, **menghapus**, **mengambil data berdasarkan index**, dan **membalik list**.

```
head
 |
 v
[10|*]--->[20|*]--->[30|*]--->[40|None]
```

Kunci semua operasi ini adalah **mengatur ulang pointer `next`** dengan urutan yang benar.

**Operasi yang dibahas:**
- `insert_after` → menyisipkan node setelah nilai tertentu
- `delete_front` → menghapus node pertama
- `delete_last` → menghapus node terakhir
- `delete_value` → menghapus node berdasarkan nilai
- `get_at` → mengambil data pada index tertentu
- `reverse` → membalik urutan list

Setiap sel di bawah memakai kelas dasar dari Pekan 2 (`init`, `is_empty`, `insert_last`, `display`) ditambah method yang sedang dibahas.

## 🔁 1. Persiapan: Kelas Dasar dari Pekan 2

### Konsep

Kita pakai kembali `Node` dan `SinglyLinkedList` dasar. Untuk latihan, kita buat list berisi `10, 20, 30, 40`.

| Baris | Penjelasan |
|---|---|
| `insert_last(x)` | Menambah node di belakang (dari Pekan 2) |
| `display()` | Menampilkan isi list (dari Pekan 2) |

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:
            cur = cur.next
        cur.next = baru

    def display(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(hasil) + " -> None")


sll = SinglyLinkedList()
for x in [10, 20, 30, 40]:
    sll.insert_last(x)
sll.display()     # Output: 10 -> 20 -> 30 -> 40 -> None
````

## ➕ 2. Menyisipkan Setelah Node Tertentu (`insert_after`)

### Konsep

Menyisipkan node baru **setelah** node yang bernilai `target`. Langkahnya:
1. Cari node `target`
2. Node baru menunjuk ke node setelah target (`baru.next = cur.next`)
3. Target menunjuk ke node baru (`cur.next = baru`)

⚠️ Urutan langkah 2 dan 3 **tidak boleh terbalik**, kalau terbalik sisa list akan terputus.

```
Sebelum insert_after(20, 25):     Sesudah:

[10]-->[20]-->[30]                [10]-->[20]-->[25]-->[30]
         ^                                  ^     ^
        cur                                cur   baru
```

| Baris | Penjelasan |
|---|---|
| `while cur is not None and cur.data != target:` | Mencari node target |
| `if cur is None:` | Target tidak ditemukan |
| `baru.next = cur.next` | Sambungkan node baru ke sisa list dulu |
| `cur.next = baru` | Baru kemudian sambungkan target ke node baru |

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:
            cur = cur.next
        cur.next = baru

    def display(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(hasil) + " -> None")

    def insert_after(self, target, data):
        cur = self.head
        while cur is not None and cur.data != target:   # cari node target
            cur = cur.next
        if cur is None:
            print(f"{target} tidak ditemukan")
            return
        baru = Node(data)
        baru.next = cur.next          # 1. node baru menunjuk node setelah target
        cur.next = baru               # 2. target menunjuk node baru


sll = SinglyLinkedList()
for x in [10, 20, 30]:
    sll.insert_last(x)

sll.insert_after(20, 25)
sll.display()             # Output: 10 -> 20 -> 25 -> 30 -> None

sll.insert_after(99, 1)   # Output: 99 tidak ditemukan
````

## ❌ 3. Menghapus Node Pertama (`delete_front`)

### Konsep

Cukup pindahkan `head` ke node kedua. Node pertama otomatis tidak terjangkau lagi (dibersihkan oleh Python). Operasi ini O(1).

```
Sebelum:  head --> [10]-->[20]-->[30]
Sesudah:  head ---------->[20]-->[30]      ([10] terlepas)
```

| Baris | Penjelasan |
|---|---|
| `if self.is_empty():` | Tidak bisa menghapus dari list kosong |
| `self.head = self.head.next` | `head` pindah ke node kedua |

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:
            cur = cur.next
        cur.next = baru

    def display(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(hasil) + " -> None")

    def delete_front(self):
        if self.is_empty():
            print("List kosong")
            return
        self.head = self.head.next    # head pindah ke node kedua


sll = SinglyLinkedList()
for x in [10, 20, 30]:
    sll.insert_last(x)

sll.delete_front()
sll.display()        # Output: 20 -> 30 -> None
````

## ❌ 4. Menghapus Node Terakhir (`delete_last`)

### Konsep

Karena tidak ada pointer mundur, kita harus berhenti di node **sebelum** yang terakhir, lalu memutus `next`-nya. Ada tiga kondisi: list kosong, list hanya satu node, dan list biasa.

| Baris | Penjelasan |
|---|---|
| `if self.head.next is None:` | Hanya 1 node: `head` menjadi `None` |
| `while cur.next.next is not None:` | Berhenti di node sebelum terakhir |
| `cur.next = None` | Memutus node terakhir |

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:
            cur = cur.next
        cur.next = baru

    def display(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(hasil) + " -> None")

    def delete_last(self):
        if self.is_empty():
            print("List kosong")
            return
        if self.head.next is None:        # hanya ada 1 node
            self.head = None
            return
        cur = self.head
        while cur.next.next is not None:  # berhenti di node SEBELUM terakhir
            cur = cur.next
        cur.next = None                   # putuskan node terakhir


sll = SinglyLinkedList()
for x in [10, 20, 30]:
    sll.insert_last(x)

sll.delete_last()
sll.display()        # Output: 10 -> 20 -> None
sll.delete_last()
sll.delete_last()
sll.display()        # Output: List kosong
````

## 🎯 5. Menghapus Berdasarkan Nilai (`delete_value`)

### Konsep

Menghapus node pertama yang bernilai `data`. Pola yang dipakai: berhenti di node **sebelum** target (`cur`), lalu lompati target dengan `cur.next = cur.next.next`.

Kasus khusus: jika yang dihapus adalah `head`, cukup pindahkan `head`.

```
delete_value(20):

[10]-->[20]-->[30]       [10]---------->[30]
  ^                        ^
 cur                      cur   ([20] terlewati)
```

| Baris | Penjelasan |
|---|---|
| `if self.head.data == data:` | Kasus khusus: target ada di `head` |
| `cur.next.data != data` | Melihat **satu node ke depan** supaya bisa berhenti sebelum target |
| `cur.next = cur.next.next` | Melompati node target sehingga terlepas dari rantai |

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:
            cur = cur.next
        cur.next = baru

    def display(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(hasil) + " -> None")

    def delete_value(self, data):
        if self.is_empty():
            print("List kosong")
            return
        if self.head.data == data:        # kasus: yang dihapus adalah head
            self.head = self.head.next
            return
        cur = self.head
        while cur.next is not None and cur.next.data != data:
            cur = cur.next                # berhenti di node SEBELUM target
        if cur.next is None:
            print(f"{data} tidak ditemukan")
        else:
            cur.next = cur.next.next      # lompati node target


sll = SinglyLinkedList()
for x in [10, 20, 30, 40]:
    sll.insert_last(x)

sll.delete_value(20)
sll.display()              # Output: 10 -> 30 -> 40 -> None
sll.delete_value(10)
sll.display()              # Output: 30 -> 40 -> None
sll.delete_value(99)       # Output: 99 tidak ditemukan
````

## 📍 6. Mengambil Data pada Index Tertentu (`get_at`)

### Konsep

Linked list tidak mendukung `list[i]`, jadi kita harus berjalan sambil menghitung index. Jika index melebihi panjang list, kembalikan `None`.

| Baris | Penjelasan |
|---|---|
| `i = 0` | Index node yang sedang dilihat |
| `if i == index:` | Ketemu posisi yang dicari |
| `return None` | Index di luar jangkauan |

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:
            cur = cur.next
        cur.next = baru

    def display(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(hasil) + " -> None")

    def get_at(self, index):
        cur = self.head
        i = 0
        while cur is not None:
            if i == index:
                return cur.data
            cur = cur.next
            i += 1
        return None                       # index di luar jangkauan


sll = SinglyLinkedList()
for x in [10, 20, 30]:
    sll.insert_last(x)

print(sll.get_at(0))    # Output: 10
print(sll.get_at(2))    # Output: 30
print(sll.get_at(5))    # Output: None
````

## 🔄 7. Membalik List (`reverse`)

### Konsep

Membalik list berarti **membalik arah semua pointer `next`**. Kita butuh tiga penunjuk: `prev` (node sebelumnya), `cur` (node sekarang), dan `nxt` (node berikutnya yang disimpan sebelum pointer dibalik).

```
Awal:    None <-- ... head --> [1]-->[2]-->[3]-->None

Langkah: [1]-->None            (balik [1])
         [2]-->[1]-->None      (balik [2])
         [3]-->[2]-->[1]-->None (balik [3])

head baru = [3]
```

| Baris | Penjelasan |
|---|---|
| `nxt = cur.next` | Simpan node berikutnya sebelum rantai diputus |
| `cur.next = prev` | Balik arah pointer node sekarang |
| `prev = cur` | Geser `prev` maju |
| `cur = nxt` | Geser `cur` maju |
| `self.head = prev` | Setelah selesai, `prev` adalah node paling belakang yang lama, jadi jadi `head` baru |

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:
            cur = cur.next
        cur.next = baru

    def display(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(hasil) + " -> None")

    def reverse(self):
        prev = None
        cur = self.head
        while cur is not None:
            nxt = cur.next                # 1. simpan node berikutnya
            cur.next = prev               # 2. balik arah pointer
            prev = cur                    # 3. geser prev
            cur = nxt                     # 4. geser cur
        self.head = prev                  # head = node yang dulu paling belakang


sll = SinglyLinkedList()
for x in [1, 2, 3, 4]:
    sll.insert_last(x)

sll.display()       # Output: 1 -> 2 -> 3 -> 4 -> None
sll.reverse()
sll.display()       # Output: 4 -> 3 -> 2 -> 1 -> None
````

## 🔑 Operasi Penting yang Wajib Dipahami

| Operasi | Cara Kerja | Kompleksitas |
|---|---|---|
| `insert_after(t, x)` | Cari target, sambung `baru.next` lalu `cur.next` | O(n) |
| `delete_front()` | `head = head.next` | O(1) |
| `delete_last()` | Berhenti di node sebelum terakhir, `next = None` | O(n) |
| `delete_value(x)` | Berhenti sebelum target, `cur.next = cur.next.next` | O(n) |
| `get_at(i)` | Telusuri sambil menghitung index | O(n) |
| `reverse()` | Balik semua pointer dengan `prev`, `cur`, `nxt` | O(n) |

> 🧠 **Tips kunci:** Hampir semua kesalahan di linked list berasal dari **urutan pointer** dan **kasus khusus**. Selalu cek: (1) list kosong, (2) list satu node, (3) target ada di `head`, (4) target ada di node terakhir. Untuk menghapus node, kamu harus berhenti di node **sebelumnya**, karena tidak ada pointer mundur.

````python
class Node:
    def __init__(self, data):
        self.data = data
        self.next = None


class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def is_empty(self):
        return self.head is None

    def insert_last(self, data):
        baru = Node(data)
        if self.is_empty():
            self.head = baru
            return
        cur = self.head
        while cur.next is not None:
            cur = cur.next
        cur.next = baru

    def display(self):
        if self.is_empty():
            print("List kosong")
            return
        cur = self.head
        hasil = []
        while cur is not None:
            hasil.append(str(cur.data))
            cur = cur.next
        print(" -> ".join(hasil) + " -> None")


def hapus_semua(sll, x):
    while sll.head is not None and sll.head.data == x:     # TODO 1
        sll.head = sll.head.next

    cur = sll.head
    while cur is not None and cur.next is not None:
        if cur.next.data == x:                             # TODO 2
            cur.next = cur.next.next
        else:
            cur = cur.next


# ===== MAIN =====
sll = SinglyLinkedList()
for x in [3, 1, 3, 2, 3, 4]:
    sll.insert_last(x)

sll.display()
hapus_semua(sll, 3)
sll.display()      # TODO 3
hapus_semua(sll, 4)
sll.display()

# Output yang diharapkan:
# 3 -> 1 -> 3 -> 2 -> 3 -> 4 -> None
# 1 -> 2 -> 4 -> None
# 1 -> 2 -> None
````
